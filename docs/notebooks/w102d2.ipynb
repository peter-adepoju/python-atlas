```python
def evaluate(tokens):
    stack = []
    for t in tokens:
        if t in "+-*/":
            b, a = stack.pop(), stack.pop()
            stack.append({"+": a + b, "-": a - b, "*": a * b, "/": a / b}[t])
        else:
            stack.append(float(t))
    return stack.pop()

print(evaluate("3 4 + 2 *".split()), evaluate("10 2 8 * + 3 -".split()))
```

```python
def max_walks(slots):
    count, free_at = 0, float("-inf")
    for start, end in sorted(slots, key=lambda s: s[1]):
        if start >= free_at:
            count, free_at = count + 1, end
    return count

print(max_walks([(1, 4), (3, 5), (0, 6), (5, 7), (8, 9), (5, 9)]), bin(40), 40 & 39, (64 & 63) == 0)
```

### Bug hunt: Subtraction the wrong way round

In reverse Polish notation, `10 3 -` means 10 minus 3, but the operands are popped into the wrong names, so this prints `-7.0`. Swap them so it prints `7.0`.

```python
def evaluate(tokens):
    stack = []
    for t in tokens:
        if t == "-":
            a, b = stack.pop(), stack.pop()
            stack.append(a - b)
        else:
            stack.append(float(t))
    return stack.pop()

print(evaluate(["10", "3", "-"]))
```

## Warm-up

Queue, stack, heap, and three bitwise operations: four structures answer "which comes next?" in four different ways (first in, last in, smallest), and bits answer questions about sets of flags in one machine operation.

## Linear structures

- **Arrays**: index in O(1); patterns: two pointers (sorted pairs, reversing), sliding windows (best stretch), prefix sums (stretch totals).
- **Linked lists**: O(1) insertion at a known node, O(n) to find it; slow and fast pointers.
- **Stacks**: most recent first: matching brackets, undo, evaluating expressions, monotonic stacks.
- **Queues** (`deque`): first in, first out: breadth-first search, buffers.
- **Hash tables** (`dict`, `set`): O(1) average lookup: counting, seen-before, grouping.

## Trees, heaps and tries

- **Binary search trees**: left smaller, right larger; search, insert and delete in O(height); in-order traversal gives sorted order.
- **Heaps** (`heapq`): the smallest in O(1), push and pop in O(log n): top k, merging, running medians, Dijkstra.
- **Tries**: a tree of characters; everything under a prefix in O(prefix length) to find.

## Graphs and strategies

- **Graphs**: adjacency lists; BFS for fewest steps, DFS for exploring and cycles, Dijkstra for weighted shortest paths, topological sort for dependencies, union-find for merging groups, Kruskal for spanning trees.
- **Greedy**: take the locally best choice when you can prove it's safe (interval scheduling by earliest end).
- **Backtracking**: try a choice, recurse, undo (combinations, word search).
- **Dynamic programming**: state, transition, base cases, order, when choices overlap and greedy fails.
- **Bits**: `n & (n - 1)` clears the lowest set bit; a power of two has exactly one bit set.
