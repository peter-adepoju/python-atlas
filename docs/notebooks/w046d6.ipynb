```python
class N:
    def __init__(self, v, l=None, r=None):
        self.v, self.l, self.r = v, l, r

def delete(n, v):
    if n is None:
        return None
    if v < n.v:
        n.l = delete(n.l, v)
    elif v > n.v:
        n.r = delete(n.r, v)
    else:
        if n.l is None:
            return n.r
        if n.r is None:
            return n.l
        s = n.r
        while s.l:
            s = s.l
        n.v = s.v
        n.r = delete(n.r, s.v)
    return n

def pre(n):
    return [] if n is None else [n.v] + pre(n.l) + pre(n.r)

root = N(8, N(4, N(2), N(6)), N(12, N(10, None, N(11)), N(14)))
print(pre(delete(root, 8)))
```

```python
class N:
    def __init__(self, v, l=None, r=None):
        self.v, self.l, self.r = v, l, r

def rotate_right(y):
    x = y.l
    y.l, x.r = x.r, y
    return x

def pre(n):
    return [] if n is None else [n.v] + pre(n.l) + pre(n.r)

print(pre(rotate_right(N(5, N(3, N(2), N(4)), N(8)))))
print(pre(rotate_right(rotate_right(N(3, N(2, N(1)))))))
```

> **Invariants.** The calendar has two rules that must hold after every operation: **no two bookings overlap**, and **the tree is an AVL tree** (ordered by start, balanced, heights correct). The first is the calendar's own rule; the second is how it stays fast. Both are **invariants**, things that are true before and after every public method, though they may be briefly false in the middle of one (half-way through a rotation, the heights are wrong). The checks today test them directly, by walking your tree. When you write a data structure, write a `check()` method that verifies every invariant, and call it in your tests after each change: it finds bugs at the operation that caused them, not three hundred operations later when an answer comes out wrong. And notice where the overlap rule is enforced: in `book`, and again in `load`, because a file is just another way for data to arrive.

### Exercise: Forge bonus: a treap

A **treap** (Raimund Seidel and Cecilia Aragon, 1989) balances itself by chance instead of by height. Each node gets a random **priority** when it's created, and the tree keeps two orders at once: it's a **BST by key**, and a **heap by priority** (every node's priority is at least as high as its children's, so the highest priority is the root). For given priorities, only one tree satisfies both, and it's the BST you'd get by inserting keys in decreasing priority order: random priorities make it a random BST (height about 3 log₂ n, Day 1) whatever order the keys actually arrive in.

- `add(key)`: insert as a leaf, as in a plain BST; then, on the way back up, while the new node's priority is higher than its parent's, rotate it up (a right rotation if it's a left child, a left one if it's a right child). With the subtree-returning style, that's one check after each recursive call. No heights to maintain.

## Forge: a booking calendar

A meeting room's calendar holds bookings, each a half-open interval of whole minutes `[start, end)` with a name: a booking from 600 to 660 and another from 660 to 720 don't clash. Bookings never overlap, so ordering them by start time orders them completely, and a balanced BST keyed by start answers every question with a floor or successor search. Use an **AVL tree** (rotations, heights in the provided `Node`), so the calendar stays fast however bookings arrive; keep a single attribute holding the root.

Build `Calendar()` with:

- `book(start, end, name)`: add the booking and return `True`, or return `False` (changing nothing) if it would overlap an existing booking. Raise `ValueError` if `start >= end`. Only the bookings just before and just after `start` can clash: check those two, not every booking.
- `cancel(start)`: remove the booking that starts at `start` and return its name; `KeyError` if none does.
- `at(t)`: the name of the booking covering minute `t`, or `None`.
- `between(t1, t2)`: every booking overlapping `[t1, t2)`, as sorted `(start, end, name)` tuples, in O(h + k): only the booking found by a floor search on `t1` can start before `t1` and still overlap.
- `next_free(t, length)`: the earliest minute `s >= t` such that `[s, s + length)` overlaps no booking.
- `bookings()`, `len(c)` and `height()` (−1 when empty).
- `save()`: JSON text `{"format": "calendar", "version": 1, "bookings": [[start, end, name], ...]}` in start order, and a class method `Calendar.load(text)` that rebuilds a calendar, raising `ValueError` (only) for anything wrong: invalid JSON, a wrong format name or version, bookings that aren't a list of three-item lists, times that aren't whole numbers (`bool` doesn't count), names that aren't strings, `start >= end`, or overlapping bookings.
