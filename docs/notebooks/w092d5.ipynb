## The shape of the talk

For a 12-minute capstone talk:

1. **Opening** (1 minute): who you are, what Waymark is, in a sentence.
2. **The problem** (1 to 2 minutes): a person and a situation. A walker on a hill an hour later than planned is more memorable than "time estimation".
3. **Demo** (3 minutes): the demo video's three tasks, live if you're confident, or the video if not. Have the video ready either way.
4. **How it works** (2 to 3 minutes): the two or three ideas that matter, with one diagram, not a tour of the code.
5. **Results** (2 minutes): the numbers from the write-up: budgets, tests, users.
6. **Close** (1 minute): what you learned, what's next, thank you.

Leave time for questions inside the slot, not after it.

## Slides

Slides are for the audience's eyes while you talk; they aren't your script. A useful rule: a title on every slide, no more than six bullets, and no more than about 40 words, with a diagram or screenshot wherever one can replace words. Put what you want to say in the **speaker notes**. Use large type (at least 24 point), high contrast, and nothing that relies on colour alone. Number your slides, so questions can refer to them.

## Rehearsing

Rehearse **out loud, standing up, timed**, at least three times, and time each section. Use the **median** of your rehearsals (Day 1 of last week again): one bad run shouldn't scare you, and one good one shouldn't reassure you. Most people run long in the demo and short in the close.

Record one rehearsal and transcribe it (free speech-to-text tools do this). From the transcript you can measure your **pace** (about 120 to 170 words a minute is comfortable to follow) and your **filler words** ("um", "uh", "basically", "you know"). A few are natural; several a minute distract. A pause is better than a filler.

## Questions

Most questions are predictable: why this approach and not another, what happens when something goes wrong, how you know it works, what you'd do differently, what was hardest. Write a short answer for each, pointing at a slide or a number where you can. When you don't know: say so, say how you'd find out, and offer to follow up. Repeat each question before answering, so the whole room (and any recording) hears it.
