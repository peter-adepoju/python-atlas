### Bug hunt: A default that was never stored

`dict.get(key, [])` returns a new empty list that isn't stored, so appending to it changes nothing and this prints `{}`. Use `setdefault` so it prints `{'Mill': ['ada', 'bo']}`.

```python
walkers_at = {}
for name, start in [("ada", "Mill"), ("bo", "Mill")]:
    walkers_at.get(start, []).append(name)
print(walkers_at)
```

### Exercise: Debugging: durations that didn't parse

`parse_duration(text)` should turn `"1h05m"`, `"45m"`, `"2h"` or `"1h 5m"` into minutes, but it only handles one shape. Rewrite it with a regular expression that allows optional hours and optional minutes (at least one), with optional spaces; raise `ValueError("bad duration: TEXT")` otherwise.

```python
def parse_duration(text):
    hours, minutes = text.split("h")
    return int(hours) * 60 + int(minutes.rstrip("m"))
```
