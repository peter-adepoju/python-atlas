> **Why not a string, or a list?.** A Python string can't change at all, so every insertion in the middle builds a whole new string: O(n) per keystroke. A list of characters is mutable, but inserting at position i shifts everything after it (Week 41): still O(n). The linked buffer pays for its O(1) keystrokes with memory (a node object per character) and with slow jumps: going to character 50,000 means walking there. Real editors often use a **gap buffer** (an array with a movable gap at the cursor) or a **rope** (a balanced tree of string pieces), which combine cheap edits with fast jumps; both are built on ideas from this phase. The linked version shows the core idea most plainly: keep the place you're working **as a reference**, and changes there cost nothing to reach.

## Forge: an editor buffer

A text editor keeps its text and a **cursor** (the point between two characters where typing goes). Almost every keystroke works at the cursor: type a character, delete the one before, move left or right. With a Python string, every insertion in the middle copies the whole text: O(n) per keystroke, which is painful for a long document. With a doubly linked list of characters and a cursor that is a **node reference**, each keystroke costs O(1).

Build `Editor`, using a circular doubly linked list with a sentinel (one node per character). The cursor is a reference to the node just **left** of the cursor position, which is the sentinel when the cursor is at the very start.

- `Editor(text="")` holds the text, with the cursor at the **end**.
- `insert(s)` inserts the characters of `s` at the cursor; the cursor ends up after them (as when typing).
- `backspace(k)` deletes up to `k` characters **before** the cursor, and returns how many it deleted.
- `left(k)` and `right(k)` move the cursor up to `k` places (stopping at the ends). Each returns the (up to) 10 characters just left of the cursor afterwards, as a string: what an editor would show beside the cursor.
- `cut(k)` removes up to `k` characters **after** the cursor (the cursor doesn't move) and keeps them as the clipboard, replacing any earlier one; it returns the cut text. `paste()` inserts the clipboard at the cursor (as `insert` does) and returns `len` of what it pasted.
- `text()` returns the whole text; `len(editor)` is its length, in O(1); `cursor_position()` is the number of characters before the cursor, kept up to date as you go (not counted by walking).
- `reverse_word()`: reverse, in place, the run of letters immediately **before** the cursor (stopping at a non-letter or the start), by relinking those nodes. The cursor stays after the run. Return the reversed run.

Every operation except `text()` must take time in proportion to the number of characters it touches (`k`, the length of the string, the run or the clipboard), never the length of the whole document: the checks run tens of thousands of operations on a long text.
