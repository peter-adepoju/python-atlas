## Warm-up

No loops: `prices * qty` multiplies matching elements, `qty >= 2` makes an array of `True`/`False` that selects elements, and `np.where` chooses between two arrays element by element. Each line runs as one compiled loop inside NumPy.

## Why vectorise?

A Python list of a million floats is a million pointers to a million separate float objects, and a loop over it runs bytecode for every element: fetch the object, check its type, look up the operation, make a new object for the result (Week 64). A NumPy **array** stores the million numbers as raw 8-byte values side by side in one block of memory, all of one type (its **dtype**). An operation such as `a * 2` runs **one** C loop over that block:

- no per-element type checks or object creation;
- memory read in order, which the processor's caches handle well;
- often **SIMD** instructions that process several numbers at once.

The style that results is **vectorisation**: express the computation as operations on whole arrays. The speed-up is large, typically 10 to 100 times, but only for code that stays inside NumPy: a Python `for` loop over a NumPy array is **slower** than over a list, because each element has to be turned into a Python object.

## Arrays

- **Creating**: `np.array([1, 2, 3])` from a list; `np.arange(start, stop, step)` like `range`; `np.zeros(n)`, `np.ones((rows, cols))`, `np.full(shape, value)`; `np.linspace(0, 1, 5)` for evenly spaced values.
- **dtype**: the element type, chosen from the data (`int64`, `float64`, `bool`) or given (`dtype=np.float32`); `a.astype(float)` converts (making a copy).
- **shape and dimensions**: `a.shape` is a tuple (`(3,)` for a vector, `(2, 3)` for 2 rows and 3 columns), `a.ndim` the number of dimensions, `a.size` the number of elements; `a.reshape(2, 3)` views the same data with a new shape (`-1` means "work it out").
- **Indexing**: `a[0]`, `a[-1]`, slices `a[1:4]`, and for 2D arrays `m[row, col]`, `m[0]` (a row), `m[:, 1]` (a column), `m[1:, :2]`.
- **Back to Python**: `a.tolist()` gives nested lists of Python numbers, and `float(x)` or `int(x)` converts a single NumPy number. (Printing NumPy numbers inside a list shows things like `np.float64(2.5)`, so convert first.)

## Element-wise operations

Arithmetic and comparisons between arrays of the same shape work **element by element**, and so do NumPy's **ufuncs** (universal functions): `np.sqrt`, `np.exp`, `np.log`, `np.abs`, `np.round`, `np.maximum(a, b)`, `np.minimum`, `np.clip(a, lo, hi)`. Comparisons give **boolean arrays**; combine them with `&` (and), `|` (or) and `~` (not), in brackets, since Python's `and`/`or` don't work on arrays.

**`np.where(condition, x, y)`** is the vectorised `x if condition else y`. Together these replace most loops with `if` statements inside.

**Aggregations** reduce an array to a number: `a.sum()`, `a.mean()`, `a.std()`, `a.min()`, `a.max()`, `a.argmin()`/`a.argmax()` (the **position** of the smallest/largest), `np.cumsum(a)` (running totals), `np.diff(a)` (differences between neighbours), `np.sort(a)`, `np.argsort(a)` (the order that would sort it), `np.unique(a)`, and counting with `np.count_nonzero(mask)` or `mask.sum()`.

## Broadcasting, masks and axes

**Broadcasting** lets arrays of **different shapes** combine. NumPy compares shapes from the **last** dimension backwards; two sizes are compatible if they're equal or one of them is 1, and a size-1 dimension (or a missing one) is stretched to match, without copying data:

- `array * 2`: the number is broadcast to every element;
- a `(3, 4)` matrix minus a `(4,)` vector: the vector is subtracted from **every row**;
- a `(3, 1)` column plus a `(1, 4)` row gives a `(3, 4)` table of all sums. `a[:, None]` (or `a.reshape(-1, 1)`) turns a vector into a column.

**Axes**: aggregations on a 2D array can work along one axis. `m.sum(axis=0)` collapses the rows, giving one total **per column**; `m.sum(axis=1)` collapses the columns, giving one total **per row**. The same goes for `mean`, `max`, `argmax` and so on.

**Boolean masks** select or change elements: `m[m < 0] = 0` sets every negative element to zero in place.

## Pitfalls

- **Slices are views, not copies.** `b = a[1:3]` shares memory with `a`: changing `b` changes `a`. That's what makes slicing fast. Call `.copy()` when you need independent data. (Boolean and list indexing, `a[mask]` and `a[[0, 2]]`, do make copies.)
- **Fixed-size integers overflow silently.** A `uint8` holds 0 to 255 and an `int8` holds -128 to 127; arithmetic beyond that **wraps around**, with no error. Python ints never overflow; NumPy ints do. Use a wide enough dtype (`int64` is the usual default) or convert first with `astype`.
- **Floats aren't exact.** `0.1 + 0.2 == 0.3` is false in NumPy too. Compare with `np.isclose(a, b)` element-wise or `np.allclose(a, b)` for the whole array.
- **`==` on arrays gives an array**, so `if a == b:` raises "the truth value of an array is ambiguous". Use `np.array_equal(a, b)`, or `.all()` / `.any()`.
- **Don't loop** over arrays element by element in Python; if you must loop, a plain list is faster.
