## What a capstone is for

A capstone does three jobs at once:

- **It proves you can build something whole.** Not an exercise with checks waiting for it, but a program with a purpose, decisions, tests, documentation and a release: the whole of Phase 6 in one place.
- **It's the centrepiece of your portfolio.** Someone deciding whether to work with you will look at one project properly. A finished, well-explained small project beats an impressive unfinished one every time.
- **It teaches what exercises can't**: scoping, changing your mind, debugging your own design, and finishing.

The six weeks are planned for you: this week, **specification** (an idea, a written spec, an architecture, a repository with CI, and a first thin slice working end to end); three **build sprints**; a week to **harden** it (performance, security, errors and logging, documentation, testing with real users); and a week to **ship** (a release, a write-up, a demo, a portfolio page). Plan on about **10 to 12 hours a week** of work on it.

**The worked example.** Alongside your own project, the course builds one in the browser, so every step has an example you can run and check: **Waymark**, a planner for walks. It holds a network of footpaths (junctions, paths between them, their length and climb), finds routes between places, estimates how long they'll take, saves favourite walks and keeps a log of walks done, with a small API and a command-line tool. It touches most of the course: classes, graphs and shortest paths, a database, an API, tests and data.

## Finding and judging ideas

Good capstone ideas usually come from **your own life**: something you or people you know do by hand, a hobby with data in it, a club that runs on spreadsheets, a tool you wished you'd had earlier in this course. Those come with real users and real data, and you'll know when the result is right. Extending an earlier project (the `spend` tracker, the `shelf` library, the helpdesk) is fine too, if the extension is substantial.

Be wary of ideas that are:

- **Too big**: "a social network", "a game engine", "my own programming language". Six weeks is about 70 hours.
- **Too vague**: "something with machine learning". An idea is a user, a problem and a thing that solves it.
- **Dependent on what you can't control**: data you can't get, an API that may change or charge, other people's time.
- **A copy of a large existing product**: the interesting parts are the bits a clone leaves out.

To choose between a few candidates, use a **weighted scoring matrix**: pick criteria (how interested you are, how useful it would be, which skills it shows, whether you can finish it, whether you can get the data), weight them by how much they matter, and score each idea from 1 to 5. Some criteria are **dealbreakers**: a 1 for feasibility or data rules an idea out whatever its total.

## How big is it?

People underestimate work, by a lot and consistently. Two habits help:

- **Three-point estimates.** For each piece of work, estimate an optimistic time `o`, a most likely time `m` and a pessimistic time `p`. The **PERT** estimate is `(o + 4m + p) / 6`, with a standard deviation of about `(p − o) / 6`. Over many tasks, the expected times **add up**, but the spreads don't: the total's standard deviation is the **square root of the sum of the variances** (each task's spread squared), because some tasks finish early while others run late.
- **Capacity, not calendar.** Ten hours a week for six weeks isn't sixty hours of building: meetings with yourself, reading documentation, being stuck and tidying up take a share. A **focus factor** of about 0.7 is realistic.

Then **cut**. Sort features with **MoSCoW**: **Must** have (without it, it isn't the product), **Should** have (important but it still works without), **Could** have (nice extras), **Won't** have this time (written down so you stop thinking about them). The **minimum viable product** (MVP) is the musts: the smallest version that's useful **end to end**. If the musts don't fit your capacity, the idea is too big: cut it, don't hope.

## Risks and the pitch

A **risk** is something that might go wrong and would hurt if it did: "I can't get footpath data", "the route search is too slow for a big map", "I lose two weeks to exams". Score each from 1 to 5 for **likelihood** and **impact**; their product ranks them. Then do something about the biggest: plan a **spike** (a short, timeboxed experiment, a few hours at most) for the riskiest unknowns **first**, so that a fatal problem shows up in week one rather than week five; keep a fallback ("use a small hand-made map").

Finally, write the **pitch**: one short paragraph saying who it's for, what problem they have, what your program does about it, and what's interesting about building it. Plain, specific words: "walkers who want a route of about two hours from their door" says more than "a seamless, innovative walking experience". If you can't write the pitch, you don't have an idea yet.

> **Waymark** helps people who walk for pleasure plan routes from a map of local footpaths: give it a start, an end or a target time, and it finds a route, estimates how long it will take including the climbing, and keeps a log of walks done. Building it means modelling a path network as a graph, searching it efficiently, storing walks in a database and serving it through a small API.
