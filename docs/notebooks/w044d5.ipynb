```python
def pair_slow(nums, target):
    checks = 0
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            checks += 1
            if nums[i] + nums[j] == target:
                return (i, j), checks
    return None, checks

def pair_fast(nums, target):
    seen = {}                                  # value -> position
    for j, x in enumerate(nums):
        if target - x in seen:
            return (seen[target - x], j), j + 1
        seen[x] = j
    return None, len(nums)

nums = list(range(0, 4000, 2))
print(pair_slow(nums, 7994), pair_fast(nums, 7994))
```

```python
def longest_run(nums):
    have = set(nums)
    best = 0
    for x in have:
        if x - 1 not in have:                  # x starts a run
            length = 1
            while x + length in have:
                length += 1
            best = max(best, length)
    return best

print(longest_run([100, 4, 200, 1, 3, 2]), longest_run([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]), longest_run([]))
```

```python
def repeats_within(nums, k):
    last = {}
    for i, x in enumerate(nums):
        if x in last and i - last[x] <= k:
            return True
        last[x] = i
    return False

print(repeats_within([1, 2, 3, 1], 3), repeats_within([1, 2, 3, 1], 2), repeats_within([1, 0, 1, 1], 1))
```

### Bug hunt: Pairing with itself

This two-sum records each number **before** looking for its partner, so a number can find itself: for target 6 it reports the 3 at position 1 paired with itself, `(1, 1)`. Look first, then record, so it prints `(1, 2)`.

```python
def two_sum(nums, target):
    seen = {}
    for j, x in enumerate(nums):
        seen[x] = j
        if target - x in seen:
            return seen[target - x], j
    return None

print(two_sum([1, 3, 3], 6))
```

### Bug hunt: Too loose a key

This groups words by the **set** of their letters, which ignores how many times each letter appears, so `"aab"` and `"abb"` are wrongly grouped as anagrams: this prints `[['aab', 'abb', 'bab']]`. Use a key that keeps the counts, so it prints `[['aab'], ['abb', 'bab']]`.

```python
def group(words):
    groups = {}
    for w in words:
        groups.setdefault(frozenset(w), []).append(w)
    return list(groups.values())

print(group(["aab", "abb", "bab"]))
```

## Have I seen its partner?

The **two-sum** pattern: for each item, ask whether its **partner** (the value that would complete the answer) has already appeared, by looking it up in a dictionary of the items seen so far. Then record the item itself. Checking **before** recording means an item can't pair with itself.

The same shape answers many questions: two numbers with a given sum or difference, a word and its reverse, a person and someone born on the same day, a transaction and its refund. The dictionary maps whatever you'll look up (the value) to whatever you need back (a position, a count, the item).

## Grouping by a canonical key

To group things that are "the same" in some sense, compute a **canonical key** (one standard form for everything in a group) and use it as a dictionary key:

| Group | Canonical key |
|---|---|
| anagrams | `"".join(sorted(word))`, or a tuple of 26 letter counts |
| words with the same letters, ignoring repeats | `frozenset(word)` |
| shifted strings (`"abc"`, `"bcd"`, `"xyz"`) | the differences between neighbouring letters, mod 26, as a tuple |
| files with the same contents | a hash of the contents (Day 1) |
| rotations of a shape | the smallest of its rotations |

Sorting each word costs O(k log k) for a word of k letters, so grouping n words is O(n k log k): far better than comparing every pair of words.

## Remembering where

A dictionary from value to its **last position** answers "how far back did I last see this?": duplicates within k positions, the longest gap between repeats, or (with the first position instead) the longest stretch between two equal values. Week 41's longest run without repeats used exactly this.

## Making keys

Dictionary keys and set items must be hashable, and lists, sets and dicts aren't (Day 1). Convert them:

| You have | Use as a key |
|---|---|
| a list | `tuple(the_list)` |
| a set | `frozenset(the_set)` |
| a dict | `frozenset(the_dict.items())` (if its values are hashable) |
| a grid (list of lists) | `tuple(map(tuple, grid))` |
| a point or pair | a tuple, `(x, y)` |
