> **One table, many answers.** The comparer builds a single LCS table and reads five different things from it: a number, a ratio, a diff, a grouped summary and marked-up text. That's typical of DP in practice. The expensive part is filling the table; walking back through it costs only O(n + m). Real tools add refinements on top: Git compares lines first and only then words within changed lines, ignores whitespace changes on request, and uses Myers' algorithm, which is fast when the two versions are mostly the same.

## Forge: a document comparer

A writing group swaps drafts and wants to see what changed between two versions, word by word. Write `DocumentCompare(old_text, new_text)`:

- Split both texts into words with `split()` (any whitespace separates words), kept as `old` and `new`. Build the LCS table over the words once, in `__init__`.
- `common()`: the number of words in a longest common subsequence.
- `similarity()`: `2 × common / (len(old) + len(new))`, rounded to 3 places; `1.0` when both are empty.
- `word_diff()`: a list of `(mark, word)` pairs, with marks `"="` (kept), `"-"` (removed) and `"+"` (added), from walking back through the table: a match goes diagonally; otherwise an addition if moving left keeps at least as much LCS as moving up, else a removal (as in Day 1).
- `changes()`: the diff grouped into runs of non-kept words, each `(kind, removed, added)` with the words joined by spaces, where kind is `"replace"` (both), `"delete"` (only removed) or `"insert"` (only added).
- `render()`: the new text with changes marked like Git's word diff: removed runs as `[-...-]`, added runs as `{+...+}` (removed first), everything separated by single spaces.
- `edit_distance()`: the word-level edit distance, with two rows.
- `shared_passage()`: the longest run of consecutive words appearing in both texts, joined by spaces (the first one found scanning the old text's end positions in order); `""` if none.
