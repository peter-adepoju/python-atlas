> **What a wrapper is for.** Look at what the wrapper hides. Callers of `parse_int` never see an end pointer; they get an `int` or a `ValueError` that says what was wrong. Callers of `sort_ints` never see a function pointer type, and can't let the callback be garbage collected mid-sort, because the wrapper owns it for the duration of the call. Callers of `Buffer` can't write past the end of the memory, because every access is checked first. That's the job of every binding layer, from `ctypes` code like this to the C extensions in the standard library: take a C API that trusts its caller completely, and turn it into a Python API that checks everything and fails with exceptions. Do it in one small module, test it thoroughly, and nothing else in the program needs to know there's C underneath.

Write a small wrapper module around the C library with `ctypes`:

- **`Library(path=None)`**: loads the library (`CDLL(path)`); `bind(name, argtypes, restype)` declares a function's types once, remembers it, and returns it; afterwards `lib.name` gives the bound function, while unbound names raise `AttributeError(f"{name} has not been bound")`; `signatures()` maps each bound name, sorted, to `"(argtype names) -> restype name"` (using each ctypes type's `__name__`, and `None` for no result).
- A module-level `libc = Library()` with `strtol`, `qsort`, `memset` and `memmove` bound.
- **`parse_int(text, base=10, strict=True)`**: uses `strtol` and its end pointer; raises `ValueError(f"no digits in {text!r}")` if nothing was used, and, when `strict`, `ValueError(f"unexpected {junk!r} after number in {text!r}")` if anything other than spaces follows the number.
- **`sort_ints(values, key=None, reverse=False)`**: sorts with `qsort` and a `CFUNCTYPE` comparison callback, by `key(value)` with ties broken by the value, descending if `reverse`, returning a new list.
- **`Buffer(size)`**: owns C memory; `write(offset, data)` (with `memmove`) and `read(offset, n)` raise `IndexError` if the range doesn't fit inside the buffer; `fill(byte)` uses `memset`; `len()`; `view()` returns a writable `memoryview` of the bytes (format `"B"`).
- **`Point`**, a `Structure` of two `c_int32` fields `x` and `y`; `pack_points(pairs)` returns a C array of `Point`s, and `unpack_points(array)` the list of `(x, y)` pairs.
