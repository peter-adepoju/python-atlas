## Values, decisions and loops

Everything is an object with a **type** (`int`, `float`, `str`, `bool`, `None`), and the type decides what operations mean (`3 * "ab"` repeats). Strings are immutable sequences: slicing, `split`, `join`, `replace`, f-strings with format specs (`f"{km:.1f}"`). Decisions use `if`/`elif`/`else` and truthiness (empty things are false). Loops: `while` for "until", `for` over any iterable, `range` for counting, `break` and `else` on loops.

## Functions and collections

Functions take positional and keyword arguments, defaults (never mutable ones), `*args` and `**kwargs`; they return one object (often a tuple). Lists (ordered, mutable), tuples (fixed), dictionaries (key to value, insertion-ordered), sets (unique, fast membership). Comprehensions build them in one expression. `sorted(items, key=...)` is stable: sort by the secondary key first, or use a tuple key.

## Files, errors and recursion

Open files with `with`, and read text with an explicit encoding. Formats: `csv` for tables, `json` for nested data (keys become strings). Catch the **specific** exceptions you expect, and raise your own with clear messages. Recursion needs a base case and progress towards it; memoise overlapping calls. Complexity: count how work grows: a loop inside a loop over the same data is O(n²); a dictionary lookup is O(1) on average; sorting is O(n log n); binary search is O(log n).

## Objects

A class bundles data and behaviour; `__init__` sets up each instance; methods take `self`. Properties validate or compute attributes. The **data model** (`__repr__`, `__eq__`, `__lt__`, `__len__`, `__iter__`, `__contains__`) makes your objects work with Python's syntax. `@dataclass` writes `__init__`, `__repr__` and `__eq__` for you (`order=True` adds comparisons, `frozen=True` makes them immutable and hashable). Inheritance shares behaviour (`super()`), abstract base classes require methods, and composition ("has a") is often better than inheritance ("is a"). Type hints document intent. Patterns such as strategy (pass a function in) keep code open to change.
