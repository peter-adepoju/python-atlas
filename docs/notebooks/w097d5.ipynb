## What they're asking

Behind each question is a **competency**: prioritising, handling feedback, communication, learning from failure, working with users, leadership. Interviewers want evidence from the past ("tell me about a time"), not promises ("I would..."). They listen for what **you** did (not "we"), how you **thought** about it, and what **changed** as a result. They also notice honesty: a failure story where you admit the mistake and show what you learned is stronger than a failure that was really someone else's fault.

## STAR

A reliable structure, kept to about two minutes (300 words at a speaking pace):

- **Situation**: a sentence or two of context.
- **Task**: what needed to happen, and your part in it.
- **Action**: what you did, step by step, and why. This is most of the answer.
- **Result**: what happened, with a number if you can (a time saved, a score that rose, a release that shipped), and what you learned.

Say "I" for your own actions. Cut hedging ("I think maybe", "sort of", "I guess"): it makes good work sound accidental.

## A story bank

Prepare six to eight stories, each tagged with the competencies it shows, and practise them aloud. Most questions can be answered from a good bank: the sprint where you cut export answers "prioritising", "deadlines" and "communication". Spread the stories, so one interview doesn't hear the same one three times.
