### Exercise: Forge: cleaning a members export

Write `clean(text)` that returns `(members, problems)`. Read the text as `;`-separated CSV with a header. For each row (`reader.line_num` gives N for problem messages), clean each field:

1. **name**: strip, collapse runs of spaces to one, and give each word a capital first letter and lower case after (`"ada  OKAFOR"` becomes `"Ada Okafor"`). Empty: problem `"line N: missing name"`.
2. **email**: strip and lower-case; must fully match the email pattern from the mixed review. Otherwise: `"line N: bad email"`.
3. **phone**: keep the digits; a leading `234` (from `+234`) becomes `0`; must then be exactly 11 digits starting with `0`. Otherwise `"line N: bad phone"`.
4. **joined**: one of three formats, `DD/MM/YYYY HH:MM`, `YYYY-MM-DD HH:MM`, or `D Mon YYYY H:MM` (like `1 Nov 2026 9:00`); a Lagos local time, converted to UTC and written as ISO with `Z` (`"2026-10-31T17:05:00Z"`). If none of the formats fits: `"line N: bad date"`.
5. **distance**: map case-insensitively `5k` to `"5K"`, `10k` to `"10K"`, `half` to `"Half marathon"`, `marathon` to `"Marathon"`. Anything else: `"line N: bad distance"`.

A row with any problem is left out of `members`, and **all** its problems are reported, in the field order above. Each member is a dictionary with keys `name`, `email`, `phone`, `joined`, `distance`. Then `to_json(text)` returns `json.dumps({"members": members, "problems": problems}, indent=2)`.

````python
import csv, io, json, re
from datetime import datetime, timezone
from zoneinfo import ZoneInfo

EMAIL = re.compile(r"[\w.%+-]+@(?:[A-Za-z0-9-]+\.)+[A-Za-z]{2,}")
LAGOS = ZoneInfo("Africa/Lagos")
FORMATS = ["%d/%m/%Y %H:%M", "%Y-%m-%d %H:%M", "%d %b %Y %H:%M"]
DISTANCES = {"5k": "5K", "10k": "10K", "half": "Half marathon", "marathon": "Marathon"}
````

````python
# Check: run the exercise cell above first, then this one.
check("TEXT = \"\"\"Name;Email;Phone;Joined;Distance\n ada  OKAFOR ;Ada.Okafor@Example.com;0803 123 4567;31/10/2026 18:05;10k\nBisi Adeyemi;bisi@mail;+234 805 987 6543;2026-11-01 07:30;half\nCHEN wei;chen.wei@club.ng;(0701) 222-3333;1 Nov 2026 9:00;5K\n;dee@club.ng;08031112222;2026-11-02 10:00;marathon\nEze Nnamdi;eze@club.ng;0803 123;2026-13-01 10:00;10K\n\"\"\"\ndef test_members():\n    \"members\"\n    members, problems = clean(TEXT)\n    assert members == [\n        {\"name\": \"Ada Okafor\", \"email\": \"ada.okafor@example.com\", \"phone\": \"08031234567\", \"joined\": \"2026-10-31T17:05:00Z\", \"distance\": \"10K\"},\n        {\"name\": \"Chen Wei\", \"email\": \"chen.wei@club.ng\", \"phone\": \"07012223333\", \"joined\": \"2026-11-01T08:00:00Z\", \"distance\": \"5K\"},\n    ]\ndef test_problems():\n    \"problems\"\n    members, problems = clean(TEXT)\n    assert problems == [\"line 3: bad email\", \"line 5: missing name\", \"line 6: bad phone\", \"line 6: bad date\"]\ndef test_json():\n    \"to_json\"\n    import json\n    data = json.loads(to_json(TEXT))\n    assert len(data[\"members\"]) == 2 and len(data[\"problems\"]) == 4\n    assert to_json(TEXT).startswith('{\\n  \"members\": [')")
````

> **Clean once, at the edge.** All the messiness was handled in one place, as the data came in. Everything after `clean` can trust the data completely: every phone has 11 digits, every time is UTC, every distance is one of four labels. That's the pattern professional systems follow: validate and normalise at the boundary, report what you can't fix (never silently drop it), and keep the rest of the program simple.

## Forge bonus: reading a calendar file

Calendar apps exchange events as **iCalendar** files (`.ics`), plain text with one property per line:

```
BEGIN:VEVENT
SUMMARY:Club run
DTSTART;TZID=Africa/Lagos:20261107T063000
DTEND;TZID=Africa/Lagos:20261107T080000
LOCATION:Lekki Phase 1
END:VEVENT
```

`DTSTART` is the start time in the basic ISO form `YYYYMMDDTHHMMSS`; `TZID=` names its time zone, or a time ending in `Z` is UTC.
