> **Key idea.** Use the object-oriented style: `fig, ax = plt.subplots()`, then draw on `ax` (`plot`, `bar`, `barh`, `scatter`, `hist`) and label it (`set_title`, `set_xlabel`, `set_ylabel`); `fig.savefig(path, dpi=..., bbox_inches="tight")` saves it. Choose by the question: line for change over time, bar for comparing categories (sorted, horizontal for long labels), scatter for relationships, histogram for distributions. Give every series a `label=` and call `ax.legend()`; style with `color`, `linestyle`, `marker` and `alpha`; add `axhline`, `annotate` and `bar_label`. `plt.subplots(rows, cols, sharey=True)` gives a grid; `fig.suptitle` and `tight_layout` finish it; pandas plots onto an axes with `.plot(kind=..., ax=ax)`. Keep charts honest: bars from zero, sorted categories, units and sources labelled, shared scales for comparisons, and colours that don't rely on red against green.

## Warm-up

`plt.subplots()` makes a **figure** (the whole picture) and one **axes** (a single chart with its own x and y axes). Everything you draw goes onto an axes: `ax.plot` adds a **line**, `ax.bar` adds one rectangle, a **patch**, per bar. Because they're objects, a program (or a check) can count and inspect them. The figure appears under the output.

## Figures and axes

matplotlib has two styles. The **object-oriented** style used here names what it draws on: `fig, ax = plt.subplots()`, then `ax.plot(...)`, `ax.set_title(...)`. The older `plt.plot(...)` style draws on "the current axes", which is quick for one chart and confusing for several. Use the axes.

The common methods on an axes:

- drawing: `ax.plot(x, y)`, `ax.bar(labels, heights)`, `ax.barh(...)`, `ax.scatter(x, y)`, `ax.hist(values, bins=...)`;
- text: `ax.set_title(...)`, `ax.set_xlabel(...)`, `ax.set_ylabel(...)`;
- limits and grid: `ax.set_ylim(bottom, top)`, `ax.grid(True)`.

`fig.savefig("chart.png", dpi=150, bbox_inches="tight")` writes the figure to a file (PNG for screens, SVG or PDF for print), and `plt.close(fig)` frees it when a program makes many charts.

After the run, `revenue.png` is listed among the files: the page shows the figure itself above that list.

## Choosing a chart

Start from the **question**, not the chart:

- **How does something change over time?** A **line** chart: time along x, the value up y. Lines imply continuity, so use them for ordered x values.
- **How do categories compare?** A **bar** chart: one bar per category, length showing the value. Sort the bars unless the categories have a natural order, and use horizontal bars (`barh`) when the labels are long.
- **Are two measurements related?** A **scatter** plot: one point per item, one measurement on each axis. A pattern rising to the right suggests they go up together.
- **How are values spread out?** A **histogram**: the range cut into bins, with a bar showing how many values fall in each. It shows the middle, the spread, the skew and the outliers that an average hides.

`axes` is a 2 × 2 array of axes, indexed like any NumPy array; `hist` returns the counts and the bin edges as well as drawing them; `tight_layout()` stops titles overlapping.

## Labels, legends and style

A chart without labels is a puzzle. Every chart needs a **title** that says what it shows, **axis labels with units** (`"Revenue (£)"`, `"Temperature (°C)"`), and, when there's more than one series, a **legend**: give each `plot` a `label=` and call `ax.legend()`.

Style arguments: `color=` (a name like `"tab:blue"` or a hex code like `"#1f77b4"`), `linestyle=` (`"-"`, `"--"`, `":"`), `marker=` (`"o"`, `"s"`, `"^"`), `linewidth=`, and `alpha=` for transparency. `ax.axhline(y)` and `ax.axvline(x)` draw reference lines (a target, an average). `ax.annotate(text, xy=(x, y), xytext=(x2, y2), arrowprops={"arrowstyle": "->"})` points at something, and `ax.bar_label(bars)` writes each bar's value on it.

## Several charts, and pandas

`plt.subplots(rows, cols, figsize=(width, height))` gives a grid of axes; `sharey=True` makes them use the same y scale (essential when the charts are meant to be compared), and `fig.suptitle(...)` titles the whole figure.

pandas draws with matplotlib directly: `series.plot(kind="bar", ax=ax)` or `df.plot(x="month", y="revenue", ax=ax)`, where passing `ax=` puts the chart on an axes you made. A `groupby` result is often exactly the Series you want to plot.

## Honest charts

The same numbers can be drawn to tell different stories. A few rules keep the story true:

- **Bars start at zero.** A bar's length is the value; cutting the axis at 95 makes 96 look twice as big as 95. (Line charts may zoom in, because they show change, but say so.)
- **Sort categories** by value unless they have a natural order (months, sizes): it makes the comparison instant.
- **Label units and sources.** "Revenue (£, thousands)" and "Source: shop orders, Jan to Apr 2026".
- **Same scale when comparing.** Side-by-side charts with different y ranges mislead; use `sharey=True`.
- **Fewer effects.** No 3D bars, no pie charts with a dozen slices, no rainbow colours. The default colour cycle (`tab:blue`, `tab:orange`, ...) is designed to be readable; avoid telling series apart **only** by red against green, which many people can't distinguish, and use markers or line styles too.
