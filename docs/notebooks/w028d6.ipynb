## Forge: a vending machine

A vending machine is a good small system: several kinds of thing, each with its own data and rules, working together. Here's the specification.

- A **`Product`** has a `name` and a `price` in pence (a positive whole number).
- A **`Slot`** has a `code` like `"A1"`, a `product`, and a `quantity` (0 to `CAPACITY`, a class attribute of 8). It can be `restock(n)`ed (never above capacity: `ValueError`) and can `take_one()` (`ValueError` when empty).
- The **`Machine`** accepts only the coins in its class attribute `COINS = (200, 100, 50, 20, 10, 5)` (pence). `Machine(float=None)` takes its starting float, a dictionary from coin values to counts (copied; empty if not given). It keeps:
  - `slots`, a dictionary from codes to slots;
  - `float`, a dictionary from coin values to how many of that coin it holds, for giving change;
  - `inserted`, the total of coins put in for the current purchase;
  - `sales`, a list of `(code, product name, price)` for every sale.

Its methods:

- `add_slot(code, product, quantity)`: a new slot (a code already used raises `ValueError`).
- `insert(coin)`: add a coin to `inserted` and to the float; any other value raises `ValueError` and isn't kept.
- `choose(code)`: buy from that slot. It returns a list of the coins given as change, largest first. It raises `KeyError` for an unknown code and `ValueError` (changing nothing) if the slot is empty, if not enough money has been inserted, or if the exact change can't be made from the float. On success: take one item, record the sale, give the change from the float, and reset `inserted` to 0.
- `cancel()`: give back exactly the coins inserted for the current purchase (largest first), taking them out of the float, and reset `inserted`.
- `takings()`: the total price of all sales.

To give change, use the **largest coin first** that fits and that the float still holds, repeatedly. (For these coin values, that always finds a way if one exists using the coins held, except in rare cases this exercise won't test.)
