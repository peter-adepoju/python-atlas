## Forge: seat booking

A concert goes on sale at 10:00 and 80,000 people try to buy 20,000 seats. The rules a booking system must never break are short: **a seat is sold at most once**, and **someone who paid gets their seats**. Everything else follows from them.

- **Holds.** Choosing seats and paying takes minutes, so the system **holds** the chosen seats for that person (ten minutes, say) and shows them to nobody else. A hold that isn't paid for **expires** on its own, so abandoned checkouts free their seats without anyone cleaning up.
- **All or nothing.** Four friends want four seats together. If any one of them is taken, the hold fails and none are held, otherwise the system fills up with useless fragments of groups.
- **Atomic claims.** Two people clicking the same seat at the same instant must not both get it. In a real system the claim is one atomic step (a conditional write, like Wednesday's rate-limiter store: "set this seat's holder only if it's free"); here, one method call does the whole check-and-claim.
- **Idempotent purchase.** The payment page times out and the browser retries. An **idempotency key** (Week 77) makes the retry return the same order instead of charging twice.

Build `BookingService(clock, hold_ttl=600, max_seats=6)`:

- `add_event(event_id, seats)` with a list of seat ids; a repeated id raises `ValueError("event exists")`.
- `hold(event_id, user, seats)` returns a hold id (`"h1"`, `"h2"`, ...). It raises `ValueError("choose 1 to N seats")` (N being `max_seats`) for an empty list or too many, `ValueError("unknown seat X")` for the first seat (in the order given) not in the event, and `ValueError("seat X is taken")` for the first seat that is sold or held by an **unexpired** hold (a user's own hold counts too). Holds expire `hold_ttl` seconds after they were made (a hold is expired when `now >= expires`). All seats are held or none.
- `release(hold_id)` cancels a hold (an unknown or finished hold is ignored).
- `purchase(hold_id, user, key)`: if `key` was used before, return the earlier order unchanged. Otherwise an unknown hold raises `KeyError("no such hold")`; another user's hold raises `PermissionError("not your hold")`; an expired or released hold raises `ValueError("hold expired")`. On success the seats are sold, the hold is finished, and it returns an order `{"order": "o1", "event": ..., "user": ..., "seats": [sorted seat ids]}` (numbered `"o1"`, `"o2"`, ...).
- `available(event_id)`: the sorted seats neither sold nor held by an unexpired hold. `stats(event_id)`: `{"sold": n, "held": n, "available": n}` (seats, not holds).

## Forge bonus: a waiting room

At 10:00 the booking service could take perhaps 2,000 buyers at once; 80,000 arrive. Letting them all in makes the site slow for everyone, and the fastest scripts win. A **virtual waiting room** puts arrivals in a first-come queue and **admits** them in batches as active buyers finish, so the service runs at a pace it can sustain and the order is fair.
