## What the page is for

A visitor to a portfolio has a question ("can this person build things like the ones we build?") and very little time. Give each project a **card**: a title, one or two sentences on what it does and for whom, a few tags (Python, pandas, A*), perhaps a screenshot, and links to the **code**, the **demo** and the **write-up**. Put the capstone first. Three to five good projects beat fifteen half-finished ones. Add a short paragraph about yourself and a way to contact you, and nothing else.

## Escaping

Generate the page from data (a list of project dicts) rather than writing HTML by hand: it's easier to keep consistent, and every value passes through one place. `html.escape(text)` turns `&`, `<`, `>` and both quote marks into entities, so it's safe inside elements and inside quoted attributes.

## Accessible by construction

The Web Content Accessibility Guidelines (WCAG, from the W3C) are long, but most failures on small sites are a handful of things a program can find:

- **Headings** form an outline: one `h1`, then `h2` for sections and `h3` within them, never skipping a level (screen-reader users jump between headings).
- **Images** need `alt` text that says what the image shows ("Elevation profile of the Ridge walk"). A file name is not a description. Purely decorative images get `alt=""` so they're skipped.
- **Links** must make sense on their own, since screen readers can list them out of context: "Source code for Waymark", not "here" or "read more". A link that opens a new tab with `target="_blank"` needs `rel="noopener"`.
- **The page** needs a language (`<html lang="en-GB">`), a viewport meta tag so it works on phones, and a meta description of up to about 160 characters for search results and link previews.

`html.parser.HTMLParser` calls a method for each start tag, end tag and piece of text, which is enough to check all of these. On your own computer, free tools such as axe and Lighthouse check much more.

## Publishing

A static page needs no server: GitHub Pages (free for public repositories) serves a folder from a repository, as this course does. Keep the page's source data in the repository, generate the HTML in CI, run the checks, and publish only if they pass.
