```python
from functools import cache

calls = 0

@cache
def fib(n):
    global calls
    calls += 1
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print(fib(10), calls, fib(90))
```

```python
def subsets(items):
    if not items:
        return [[]]
    rest = subsets(items[1:])
    return rest + [[items[0]] + s for s in rest]

def perms(text):
    if len(text) <= 1:
        return [text]
    return [text[i] + p for i in range(len(text)) for p in perms(text[:i] + text[i + 1:])]

print(subsets(["a", "b"]))
print(perms("abc"))
```

> **Key idea.** Several recursive calls make a tree; repeated subproblems make it exponential, and memoisation (`@cache`) removes the repeats. Subsets come from "in or out" for each item; permutations from "each item first, then the rest".

`power(2, 10)` works, but `power(2, 5)` crashes with `RecursionError`. Fix it so it works for every exponent of 0 or more, keeping the halving idea (it makes only about log₂(e) calls).

```python
def power(b, e):
    if e == 0:
        return 1
    half = power(b, e // 2)
    if e % 2 == 0:
        return half * half
    return b * power(b, e - 2) * b
```

### Bug hunt: Leftovers

`leaves(tree)` collects the non-list values from a nested list, in order. The first call is right, but the second call returns the first call's leaves too. Fix it, keeping the `out` parameter.

```python
def leaves(tree, out=[]):
    for item in tree:
        if isinstance(item, list):
            leaves(item, out)
        else:
            out.append(item)
    return out

print(leaves([1, [2, [3]]]))
print(leaves([4]))
```

## Recursion: the essentials

A recursive function solves a problem by calling itself on a **smaller** version of the same problem. Every correct one has:

- a **base case**: an input small enough to answer directly, without recursion;
- a **recursive case** that makes **progress**: each call moves closer to the base case;
- **trust**: assume the smaller call returns the right answer, and build yours from it.

Each call gets its own **frame** on the call stack, with its own variables. Calls go **down** until a base case returns, then the answers come back **up**, each waiting call finishing its own work with the value it gets.

Code **before** the recursive call runs on the way down; code **after** it runs on the way back up, in reverse order.

Python stops recursion that goes about **1000** calls deep with `RecursionError`. That's why a loop is usually better for simple counting down a long list, and recursion is best where the problem itself branches or nests (Week 15, Day 5).

## Recursion on data

Recursion fits data that contains smaller versions of itself: a list is its first item plus a smaller list; a nested list contains lists; a folder contains folders.

**Divide and conquer** splits the problem into halves, solves each half, and combines the answers: merge sort (Week 16, Day 1, and Week 19) is the famous example. Halving means only about log₂(n) levels of calls, so it never gets near the recursion limit.

## Tree recursion and memoisation

When a function calls itself **more than once**, the calls form a tree, and the number of calls can grow exponentially:

177 calls to work out the 10th Fibonacci number, because the same small calls are repeated again and again. **Memoisation** stores each answer the first time it's worked out, and looks it up after that. `functools.cache` does it for you:

Tree recursion also **generates** things: every subset (each item is either in or out: 2ⁿ of them) or every permutation (each item can come first, then permute the rest: n! of them).

## The classic mistakes

1. **No base case, or one that's never reached**: `RecursionError`. Check that every path makes progress towards it: `n - 2` from an odd number jumps straight past a base case of `n == 0`.
2. **Forgetting to return** the recursive call's result: `f(n - 1) + n` without `return` gives `None`.
3. **A mutable default argument** used as an accumulator (`def walk(x, out=[])`): the same list is shared by every call to the function, so results pile up from one call to the next. Use `out=None` and create the list inside.
4. **Ignoring the result** of a recursive call that returns a new value: `flatten(item)` on its own line does nothing useful.
5. **Exponential tree recursion** where memoisation, or a loop, is needed.
6. **Slicing in every call** on very long lists: `xs[1:]` copies the list each time, making a simple recursive sum O(n²). Fine for exercises; pass an index instead when it matters.
