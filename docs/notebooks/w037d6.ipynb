> **A failure for every layer.** Look at where each kind of failure ends up. A broken amount becomes a `ValueError` in `parse_amount`, is translated into a `RowError` with a line number in `parse_row` (keeping the original as its cause), is logged once as a warning and collected by the importer, and reaches the user as one line under "problems:". A missing file becomes a `SourceError` at the reading layer and an exit code of 2 at the front end. Too much bad data becomes `TooManyErrors`, carrying every row's error in a group. And a bug, such as a `TypeError`, is caught nowhere, so it crashes with a full traceback and gets fixed. Each layer handles what it understands and translates the rest: that's what makes a program robust rather than merely quiet.

### Exercise: Forge bonus: retries and a circuit breaker

Programs that call other services over a network need two classic defences:

**Retrying.** Write `retry(func, attempts, sleep, log, retry_on=(ConnectionError, TimeoutError), first_delay=1)`. It calls `func()` and returns its result. If it raises one of `retry_on`, it logs a `WARNING` `"attempt N of A failed: MESSAGE"`, calls `sleep(delay)` and tries again; the delay starts at `first_delay` and doubles each time. It never sleeps after the last attempt. If every attempt fails, it raises `ExceptionGroup(f"all {attempts} attempts failed", errors)`. An exception **not** in `retry_on` goes straight through on the first occurrence. `attempts` below 1 raises `ValueError`. (`sleep` is passed in, as in Week 36, so the checks don't have to wait.)

**A circuit breaker.** When a service is down, retrying every call makes things worse. Write `CircuitOpen(Exception)` and `CircuitBreaker(threshold, cool_down)` with a `state` attribute (`"closed"`, `"open"` or `"half-open"`) and `call(func)`:

- **closed** (the start): call `func()`. A success resets the count of failures in a row; a failure adds one, and at `threshold` failures in a row the breaker becomes **open**. The failure's exception is re-raised either way.
- **open**: refuse the next `cool_down` calls without calling `func`, raising `CircuitOpen("circuit open")`. The call after those becomes a trial: the state becomes **half-open** and `func` is called.
- **half-open**: a success closes the breaker (failures back to 0); a failure opens it again at once (and its exception is re-raised).

## Forge: a statement importer

A bank lets people download their statements as text, one transaction per line, `date|description|amount`:

```text
# March
2026-03-01|Rent|-750.00
2026-03-02|Salary|£1,850.00
2026-03-02| Coffee |-2.8
```

Real files are messy: blank lines, comments, pound signs and commas in amounts, and the odd broken row. The importer should accept reasonable variations, report every broken row precisely, give up if there are too many, and log what it does. Here's the specification, from the bottom up.

**Errors.** `LedgerError(Exception)` is the base. `SourceError(LedgerError)` is for a file that can't be read. `RowError(LedgerError, ValueError)` has `__init__(self, line_no, reason)`, stores both, and has the message `"line N: reason"`. `TooManyErrors(LedgerError)` is for giving up.

**Domain.** `Transaction`, a **frozen** dataclass with `date` (a `datetime.date`), `description` (text) and `pence` (an int).

**Parsing** (generous at the edge, strict inside):

- `parse_amount(text)` returns pence. It strips spaces, removes commas, and accepts an optional sign (`-` or `+`) followed by an optional `£`, then whole pounds and optionally a `.` with one or two digits: `"-£3.20"` is `-320`, `"£1,850"` is `185000`, `"12.5"` is `1250`. Anything else (`""`, `"£"`, `"1.234"`, `"1."`, `".5"`, `"ten"`) raises `ValueError`.
- `parse_date(text)` returns a `datetime.date` from exactly `YYYY-MM-DD` (spaces around are allowed). Anything else raises `ValueError`, including impossible dates like `2026-02-30` (let `datetime.date` decide those).
- `parse_row(line_no, line)` splits on `|` and returns a `Transaction`, or raises `RowError` with one of these reasons: `"expected 3 fields, got K"`; `"bad date 'TEXT'"` and `"bad amount 'TEXT'"` (the stripped text, chained **from** the `ValueError`); `"no description"` (the description is stripped); `"amount is zero"`.

**Reading.** `read_lines(path)` returns the file's lines (UTF-8, without line endings), or raises `SourceError(f"can't read {path}")` chained from the `OSError` or `UnicodeDecodeError`.

**The importer.** `Report` is a frozen dataclass with `transactions` and `errors` (both tuples; the errors are the `RowError` messages) and a `balance` property, the sum of the pence. `Importer(log, max_errors=5)` has `run(path)`, which returns a `Report`:

- It skips blank lines and lines starting with `#` (after leading spaces). Each other line is a **row**.
- A row identical to an earlier transaction (same date, description and amount) is a `RowError` with the reason `"duplicate of line M"`.
- Each bad row is logged as a `WARNING` with its message and collected. As soon as there are **more** than `max_errors`, it raises `TooManyErrors(f"more than {max_errors} bad rows")` chained **from** an `ExceptionGroup("bad rows", ...)` of all the `RowError`s so far.
- Each good row logs `DEBUG` `"line N ok"`. At the end it logs `INFO` `"imported G of R rows"` (good rows, all rows).

**Presentation.** `money(pence)` formats pence as `£1,850.00` or `-£2.80`. `main(path, log, max_errors=5)` runs an importer and returns `(exit_code, text)`:

- Success: the text is `"N transactions, balance MONEY"`, followed (if there were bad rows) by a line `"problems:"` and one line per error, each starting with two spaces. The exit code is 0 with no problems, 1 with some.
- A `SourceError` or `TooManyErrors`: the text is `"error: "` and the message, the exit code is 2, and `main` logs it once as an `ERROR`. Anything else (a bug) isn't caught.
