### Bug hunt: C6: Three bugs

`cheapest_network(n, cables)` should return the least total cost of cables `(a, b, cost)` connecting all n buildings (Kruskal's algorithm), or `None` if they can't all be connected. It has three bugs: it sorts the cables by their first building instead of their cost, it joins the two buildings themselves instead of their sets' roots, and it never checks whether everything ended up connected. Fix them so this prints `[6, None]`.

```python
def cheapest_network(n, cables):
    parent = list(range(n))
    def find(x):
        while parent[x] != x:
            x = parent[x]
        return x
    total = 0
    for a, b, cost in sorted(cables):
        if find(a) != find(b):
            parent[a] = b
            total += cost
    return total

cables = [(0, 1, 4), (1, 2, 1), (0, 2, 3), (2, 3, 2), (1, 3, 5)]
print([cheapest_network(4, cables), cheapest_network(5, cables)])
```
