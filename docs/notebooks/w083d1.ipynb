## Warm-up

A two-dimensional array is a table: here, three students (rows) and three tests (columns). `scores[1]` is a row, `scores[:, 2]` a column. **axis=0** collapses the rows, giving one answer per column (each test's average); **axis=1** collapses the columns, one answer per row (each student's best). Most of data work is choosing the right axis.

## Building arrays

Arrays rarely come from typing lists. NumPy builds them from a description:

- `np.arange(start, stop, step)`: like `range`, but an array (and steps may be floats, though `linspace` is safer for those);
- `np.linspace(start, stop, num)`: `num` evenly spaced values **including** both ends;
- `np.zeros(shape)`, `np.ones(shape)`, `np.full(shape, value)`: filled arrays of a given shape (a tuple for more than one dimension);
- `a.reshape(rows, cols)`: the same values in a new shape (`-1` means "work it out").

Every array has a **shape** (a tuple of sizes), an **ndim** (the number of dimensions), a **size** (the number of elements) and a **dtype**, the one type all its elements share. `np.array([1, 2, 3])` holds integers; one float makes them all floats.

`.tolist()` turns an array back into plain Python lists, which print the same everywhere; this page uses it so that outputs are easy to read.

## Indexing tables

Two-dimensional indexing takes a **row** part and a **column** part, separated by a comma:

- `a[r, c]`: one element; `a[r]` or `a[r, :]`: row `r`; `a[:, c]`: column `c`;
- `a[0:2, 1:]`: a block (slices in both directions); slices are **views** that share memory (Week 65), so copy with `.copy()` before changing one;
- `a[mask]`: with a boolean array of the same shape, the elements where it's `True`, as a flat array; a mask of **rows** (one `True`/`False` per row) picks whole rows: `a[a[:, 0] > 60]`;
- `a[[2, 0]]`: **fancy indexing** with a list of positions picks those rows in that order (always a copy).

The last line combines everything: `scores >= 60` is a table of booleans, `.all(axis=1)` asks "is every test passed?" for each row, and the resulting mask picks names.

## Statistics along an axis

`sum`, `mean`, `median` (as `np.median(a, axis=...)`), `std`, `min`, `max`, `argmin` and `argmax` (the **position** of the smallest or largest) all take an `axis`. `cumsum` gives running totals; `np.diff` the change from each element to the next; `np.percentile(a, q)` the value below which `q`% of the data lies.

To use a per-row or per-column result back on the table, it needs a shape that broadcasts. A column's means have shape `(cols,)`, which broadcasts across rows automatically; a row's means have shape `(rows,)`, which does **not** line up with the columns. **`keepdims=True`** keeps the collapsed axis with size 1, giving `(rows, 1)`, which does.

`centred` shows each day relative to its own city's average: the same calculation for every row, written once.

## Missing values, sorting and counting

Real data has gaps. NumPy marks a missing float as **`np.nan`** ("not a number"), which spreads: any sum or mean involving it is `nan`. And `nan` isn't equal to anything, not even itself, so `a == np.nan` finds nothing: use **`np.isnan(a)`**. The `nan` versions of the statistics (`np.nansum`, `np.nanmean`, `np.nanmax`, ...) skip the gaps. `np.where(np.isnan(a), fill, a)` fills them.

Integers have no `nan`, so a column with gaps is always floats.

For order: `np.sort(a)` returns the sorted values; **`np.argsort(a)`** returns the **positions** that would sort them, which you use to sort something else (names by their scores) the same way. `np.unique(a, return_counts=True)` gives the distinct values and how often each appears, and `np.bincount` counts small non-negative integers.

## Random numbers

For simulations and samples, create a **Generator** with a **seed**: `rng = np.random.default_rng(42)`. The same seed gives the same numbers every time, so results can be checked and repeated; pass the generator into functions rather than using a global one. It has `rng.integers(low, high, size)` (high excluded), `rng.random(size)` (floats in [0, 1)), `rng.normal(mean, sd, size)`, `rng.choice(a, size, replace=False)` and `rng.permutation(a)`.

The exact numbers a seed produces can change between NumPy versions, so tests compare against a second generator with the same seed, or check properties (the range, the mean), rather than writing the numbers down.
