## Warm-up

205 hours of topics, one that needs three others first, and 20 days of practice in the last 25. At four hours a week, 205 hours is about a year. That's the point of today: turning a wish list into a plan that fits a real life.

## Directions

After two years you know the language well and you've touched most of what it's used for. The next step is **depth in one or two directions**, chosen by what you enjoyed and what you want to do:

- **The web**: a framework in depth, databases, deployment, a service people use.
- **Data and machine learning**: statistics, a proper course in machine learning, larger datasets.
- **Tools and the language itself**: packaging, type checking, CPython's internals, contributing to the projects you use.
- **Teaching and community**: tutorials, mentoring, talks, maintaining an open-source project.

Keep one thing broad: **reading other people's code**. It's the habit that grows you fastest in any direction.

## A plan you can keep

List the topics you want, how many hours each needs, and what each needs first. Order them so nothing comes before what it depends on (a **topological order**, from Phase 4, with alphabetical order breaking ties so the plan is stable). Then fit the order to your **weekly hours**: an honest number, such as four hours a week, not the twenty you'd manage in a perfect week. A topic can run across several weeks; what matters is the week each one finishes, so you can see the whole year.

## Goals and review

A useful goal says **what** you'll do, **how much** (a number) and **by when** (a date): "finish the machine learning course and train one model on my own data by 30 June". "Get better at Python" can't be checked, so it can't be met.

And keep what you learn: **spaced review** (revisiting a topic after 1, 3, 7, 14 and 30 days) holds onto far more than one long session does. Your course exercises are ready-made review material: in a month, redo the ones you found hard.

## Habits

A habit is easier to keep when you can see it. Track the days you practise, and notice two numbers: your **current streak** (consecutive days up to today) and your **longest**. Don't chase a perfect streak: a missed day is normal; two missed days in a row is the thing to avoid. Small and regular beats large and rare.
