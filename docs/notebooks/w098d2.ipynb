### Bug hunt: Duplicate combinations

Each recursive call restarts from the first candidate, so `[2, 3, 3]`, `[3, 2, 3]` and `[3, 3, 2]` are all counted, and this prints `4` for target 8 with `[2, 3]` where the answer is `2` (`[2, 2, 2, 2]` and `[2, 3, 3]`). Start from the current candidate so it prints `2`.

```python
def count(candidates, remaining, start=0):
    if remaining == 0:
        return 1
    total = 0
    for i in range(len(candidates)):
        if candidates[i] <= remaining:
            total += count(candidates, remaining - candidates[i], i)
    return total

print(count([2, 3], 8))
```

### Bug hunt: A room that never freed up

A room whose meeting ends at 5 is free for one starting at 5, but `<` treats them as clashing, and this prints `2`. Fix it so it prints `1`.

```python
import heapq

def rooms_needed(meetings):
    ends = []
    for start, end in sorted(meetings):
        if ends and ends[0] < start:
            heapq.heapreplace(ends, end)
        else:
            heapq.heappush(ends, end)
    return len(ends)

print(rooms_needed([(1, 5), (5, 8)]))
```

## Warm-up

`tails[i]` is the smallest possible last value of an increasing run of length `i + 1` seen so far. Its length is the length of the longest increasing subsequence (here 4, for example 1, 4, 5, 9), found with one binary search per number. The list itself isn't a subsequence of the input; only its length means anything.

## Harder patterns

- "How many at once?" (rooms, servers, overlapping walks): sort by start and keep a **min-heap of end times**; the heap's size at its largest is the answer.
- "All combinations or arrangements": **backtracking**: choose, recurse, un-choose. Avoid duplicates by only choosing candidates from the current index onwards.
- "Longest increasing...": the `tails` trick above, O(n log n), instead of the O(n²) table.
- "Fewest edits from one string to another": a table `d[i][j]` of the edit distance between prefixes, filled from the top left: `d[i][j] = d[i-1][j-1]` if the characters match, otherwise 1 plus the smallest of the three neighbours. This is Levenshtein distance, after Vladimir Levenshtein.
- "Shortest time through a weighted network": **Dijkstra's algorithm** with a heap (Phase 4's Week 49).
