````python
def sift_down(a, i, n):
    """Max-heap sift-down within a[:n]."""
    while True:
        largest = i
        for child in (2 * i + 1, 2 * i + 2):
            if child < n and a[child] > a[largest]:
                largest = child
        if largest == i:
            return
        a[i], a[largest] = a[largest], a[i]
        i = largest

def heap_sort(a):
    n = len(a)
    for i in range(n // 2 - 1, -1, -1):      # phase 1: build a max-heap, O(n)
        sift_down(a, i, n)
    print("heap:", a)
    for end in range(n - 1, 0, -1):          # phase 2: n - 1 swaps and sifts, O(n log n)
        a[0], a[end] = a[end], a[0]
        sift_down(a, 0, end)
        print(f"{a[:end]} | {a[end:]}")

values = [5, 2, 9, 1, 7]
heap_sort(values)
print(values)
````

````python
import heapq

def sort_k_sorted(items, k):
    heap = list(items[:k + 1])
    heapq.heapify(heap)
    out = []
    for x in items[k + 1:]:
        out.append(heapq.heappushpop(heap, x))   # push the next, pop the smallest
    while heap:
        out.append(heapq.heappop(heap))
    return out

readings = [3, 1, 2, 6, 4, 5, 9, 7, 8]       # each at most 2 places out
print(sort_k_sorted(readings, 2))
````

> **Key idea.** Heap sort builds a max-heap in place (O(n)), then n − 1 times swaps the root with the last heap item, shrinks the heap by one, and sifts down within it: O(n log n) worst case, O(1) extra space, not stable, and slower in practice than merge sort, quicksort or Timsort because it jumps around memory. Introsort uses it as a fallback. Make any sort stable by sorting on (key, position). Heapify plus k pops gives the k smallest in order in O(n + k log n). A k-sorted list sorts in O(n log k) with a heap of k + 1 items and `heappushpop`.

Sift within the shrinking heap only, so it prints `[1, 2, 5, 7, 9]`.

````python
def sift_down(a, i, n):
    while True:
        largest = i
        for child in (2 * i + 1, 2 * i + 2):
            if child < n and a[child] > a[largest]:
                largest = child
        if largest == i:
            return
        a[i], a[largest] = a[largest], a[i]
        i = largest

def heap_sort(a):
    n = len(a)
    for i in range(n // 2 - 1, -1, -1):
        sift_down(a, i, n)
    for end in range(n - 1, 0, -1):
        a[0], a[end] = a[end], a[0]
        sift_down(a, 0, n)

values = [5, 2, 9, 1, 7]
heap_sort(values)
print(values)
````

## Sorting in place

Heap sort has two phases, both on the one list:

1. **Build a max-heap** of the whole list, bottom-up (Day 1's heapify, with the comparisons flipped). Now the largest value is at index 0.
2. **Repeatedly**: swap index 0 (the largest remaining) with the **last item of the heap**, then shrink the heap by one, so that item is now outside it, in its final place; then sift the new root down **within the smaller heap**.

The list is always split in two: a max-heap at the front, and a sorted tail at the back holding the largest values in order.

```text
max-heap [9 7 5 1 2] |          swap 9 to the end, sift down within the first 4:
         [7 2 5 1] | 9          swap 7 to the end, sift down within the first 3:
         [5 2 1] | 7 9          ...
         [1] | 2 5 7 9          done: [1, 2, 5, 7, 9]
```

The sift-down needs a size limit `n`, so it ignores the sorted tail; that's the only change from Day 1.

## How it compares

| | worst time | extra space | stable | in practice |
|---|---|---|---|---|
| merge sort (Week 19) | O(n log n) | O(n) | yes | steady |
| quicksort (Week 19) | O(n²) | O(log n) | no | usually fastest |
| heap sort | O(n log n) | O(1) | no | slower: jumps around memory |
| Python's `sorted` (Timsort) | O(n log n) | O(n) | yes | very fast on partly sorted data |

Heap sort's guarantees are excellent, but it's usually slower than the others in practice: each sift-down jumps from i to 2i + 1, far away in memory once the heap is large, and computers are fastest reading memory in order. It also does about 2n log₂ n comparisons, twice what merge sort needs (today's challenge cuts that down).

Where it shines is as a safety net. **Introsort** (David Musser, 1997) runs quicksort, but switches to heap sort if the recursion gets suspiciously deep, so it keeps quicksort's speed and heap sort's O(n log n) worst case. C++'s `std::sort` is usually an introsort. And heap sort needs no extra memory, which matters on small devices.

In your own Python code, use `sorted` or `list.sort`. You write heap sort to understand it, and because its pieces (heapify, the shrinking boundary, partial sorting) turn up elsewhere.

## Not stable

A sort is **stable** (Week 19) when items with equal keys keep their original order. Heap sort isn't: swapping the root to the far end of the list leapfrogs it over other items with the same key.

To make any unstable sort stable, sort by `(key, original_position)` instead: no two items tie any more, and among equal keys the earlier position wins.

## Partial sorts

Often you need only the **smallest k** values, in order: the top ten results, the next five appointments. A full sort is O(n log n). With a heap:

- **Heapify** everything into a min-heap: O(n).
- **Pop k times**: O(k log n).

Total O(n + k log n), which is O(n) when k is small. That's one way `heapq.nsmallest` could work; tomorrow you'll meet another, which uses a heap of only k items and so suits data too big to hold.

Stopping heap sort part-way gives the same result in place: after j rounds of phase 2, the last j places hold the j largest values in order.

## Nearly sorted data

A list is **k-sorted** when every item is at most k places from where it belongs in sorted order: readings from sensors that arrive slightly out of order, or log lines from several machines whose clocks differ a little. The smallest item must be somewhere in the first k + 1 places, so:

- keep a min-heap of the next k + 1 items;
- pop the smallest (it's the next item of the output) and push the next item from the input.

Each step is O(log k), so the whole sort is O(n log k), better than O(n log n) when k is small.

`heappushpop` is the right tool here: pushing first means a new item smaller than everything in the heap comes straight back out.
