### Bug hunt: The empty case

This returns the largest value in a list, but crashes on an empty one. The question says to return `None` then. Make it print `[9, None]`.

```python
def largest(nums):
    best = nums[0]
    for x in nums[1:]:
        best = max(best, x)
    return best

print([largest([3, 9, 2]), largest([])])
```

### Bug hunt: Changing the caller's list

This finds the median by sorting the list **in place**, which reorders the caller's data, so the second print shows the list changed: `[1, 2, 3, 5, 9]`. Sort a copy so it prints the original order `[5, 1, 9, 3, 2]`.

```python
def median(nums):
    nums.sort()
    return nums[len(nums) // 2]

data = [5, 1, 9, 3, 2]
print(median(data))
print(data)
```

## Edge-case habits

Before pressing Check on any exam problem, run these in your head or in code:

- **Empty** input (and a single item).
- **All equal** values, and ties: does your tie rule match the question?
