> **Two dimensions from two constraints.** The festival table has one dimension for each thing the question constrains: which acts have been considered (because of the time slots) and how much budget is left. That's the general rule for designing a DP state when a problem has several limits: each independent limit becomes a coordinate, and the table's size is the product of their ranges. It's why a third limit (say, a cap on the number of acts) is affordable when its range is small, and why limits measured in millions make DP tables impossible, sending you back to last week's branch and bound.

## Forge: a festival planner

A town festival has one stage and a budget for fees. Each act has a fee, an expected crowd, and a time slot from `start` to `end` (whole hours; an act ending at 14 and one starting at 14 don't clash). The organisers want the biggest total crowd from acts that don't overlap, within the budget.

It's two of this week's ideas at once. Sort acts by **end time**: the weighted interval schedule (Week 52 Day 2's challenge) says f(i) chooses between leaving act i out and taking it, which jumps back to the last act that ends by its start. The budget adds the knapsack's second dimension.

Write `FestivalPlanner(acts, budget)`, where `acts` maps a name to `(fee, crowd, start, end)`:

- Raise `ValueError` for a negative budget, fee or crowd, or an act whose end isn't after its start.
- `best_lineup()`: `(crowd, names)`, the largest total crowd from non-overlapping acts whose fees fit the budget, with the names in order of start time. Use a table best(i, b) over the first i acts in end order and budget b, with `bisect` to find how many acts end by each act's start; walk back to find the acts.
- `best_unlimited()`: the largest crowd ignoring the budget (one-dimensional).
- `cheapest_for(target)`: the least total fee for non-overlapping acts with a total crowd of **at least** `target`, or `None`. Swap the roles: a table of the least fee over (acts considered, crowd still needed), where the crowd still needed never drops below 0.
- `count_lineups()`: how many sets of non-overlapping acts (including the empty set) fit the budget.
