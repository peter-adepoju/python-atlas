### Bug hunt: A millisecond short

Cutting the fraction of a second with `int` turns 2.3 s into `00:00:02,299`. Round to the nearest millisecond instead, so it prints `00:00:02,300`.

````python
def srt_time(seconds):
    whole = int(seconds)
    ms = int((seconds % 1) * 1000)
    h, rest = divmod(whole, 3600)
    m, s = divmod(rest, 60)
    return f"{h:02d}:{m:02d}:{s:02d},{ms:03d}"

print(srt_time(2.3))
````

## Warm-up

`2.3` isn't exactly 2.3 in binary, so `seconds % 1` is a hair under 0.3 and `int` cuts it to 299 milliseconds. Caption times are written to the millisecond, so they need rounding, not cutting.

## Timing the script

People speak about **150 words a minute** when explaining something clearly. So a scene's length can be estimated before recording: words divided by 150, times 60 seconds. A 63-word scene takes about 25 seconds; if its budget is 20, cut words now rather than talking faster later.

## Captions

**SRT** is the simplest caption format, accepted almost everywhere. Each caption is a block: its number, a time range, and one or two lines of text, with a blank line between blocks:

```
1
00:00:00,000 --> 00:00:03,500
Waymark plans walks with honest times,

2
00:00:03,500 --> 00:00:06,000
counting the climbing as well as the distance.
```

Times are `HH:MM:SS,mmm`, with a **comma** before the milliseconds. Common guidelines (the BBC's subtitle guidelines are a good free reference) say: at most two lines, about **42 characters** a line, on screen for at least a second, no overlaps, and no more than about **17 characters a second** of reading.

## Chapters and recording

Video sites turn lines like `0:00 Intro` in a description into **chapters**. YouTube's rules: the first must start at `0:00`, there must be at least three, and each must be at least 10 seconds long. Generate them from the script's timings, then adjust after recording.

Record with a free tool such as OBS Studio, in short takes (one per scene) so a mistake costs one scene; then cut, add the captions file, and watch it once with the sound off.
