### Bug hunt: A window that moved backwards

In `"abba"`, the second `a` was last seen at index 0, before the window starts, but this moves `start` back to 1 and reports `3`. Only move `start` forwards, so it prints `2`.

```python
def longest_unique(s):
    last, start, best = {}, 0, 0
    for i, ch in enumerate(s):
        if ch in last:
            start = last[ch] + 1
        last[ch] = i
        best = max(best, i - start + 1)
    return best

print(longest_unique("abba"))
```

### Bug hunt: A stack left half full

`"(("` never closes its brackets, but this only checks closing brackets and prints `True`. Check the stack is empty at the end so it prints `False`.

```python
def balanced(s):
    stack = []
    for ch in s:
        if ch == "(":
            stack.append(ch)
        elif ch == ")":
            if not stack:
                return False
            stack.pop()
    return True

print(balanced("(("))
```

## The method

Interviewers want to see how you think as much as whether you finish. A method keeps you calm and shows your thinking:

1. **Clarify** (2 minutes): restate the problem; ask about input sizes, empty inputs, duplicates, negative numbers, what to return when there's no answer.
2. **Examples**: work one small example by hand, and one edge case. Say them aloud.
3. **A simple solution first**: describe the brute force and its complexity ("checking every pair is O(n²)"). It proves you understand the problem, and it's a fallback.
4. **Improve it**: what work is repeated? Could a dictionary, a sort, two pointers or a window remove it? Say the new complexity before coding.
5. **Code** it cleanly, talking as you go: good names, small helper functions.
6. **Test** by tracing your example through the code, then the edge cases. Find your own bugs before the interviewer does.
7. **Complexity**: time and space, with a reason.

If you're stuck, say what you're thinking; interviewers give hints to people who show their reasoning.

## Patterns to look for

- "Find a pair (or a complement)": a **dictionary** of what you've seen, O(n).
- "Longest or shortest stretch with a property": a **sliding window** with two indices.
- "Sorted input" or "pairs from both ends": **two pointers**.
- "Matching, nesting, most recent first": a **stack**.
- "Overlapping ranges": **sort by start**, then sweep.
- "Top k" or "most frequent": **count, then sort or use a heap**.
