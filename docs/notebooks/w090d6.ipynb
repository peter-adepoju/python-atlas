## What a freeze means

A **feature freeze** is a promise: from today, changes fix, harden or document what exists, and nothing new goes in. It's how a project stops moving long enough to be made reliable. Before freezing, check every story:

- **Done and verified**: it has passing tests, no failing ones, and a demo step that passes. Only these count as finished.
- **Done but unverified**: someone says it's done, but a test fails, there are no tests, or the demo doesn't show it. These block the freeze, because "done" without evidence is a guess.
- **Open**: a **could** that's open simply moves to "later" (a future version). An open **must** or **should** blocks the freeze: either finish it, or cut it openly and tell everyone.
- **Cut**: listed, so nobody wonders where it went.

## Bonus: the changelog

A **changelog** tells users what changed, version by version, in their words rather than the code's. Keep a Changelog (keepachangelog.com, by Olivier Lacan) suggests the format most projects use: newest first, grouped under headings. Waymark's is written from the same story records as the freeze report, so it can't disagree with them.
