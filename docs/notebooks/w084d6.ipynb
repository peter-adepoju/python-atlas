## Forge: a spam filter

A model needs numbers, and a message is words. The standard first step is the **bag of words**: split each message into **tokens** (here, runs of lower-case letters, digits and `£`), build a **vocabulary** of the tokens seen in the training messages, and represent each message by **how many times** it contains each vocabulary word. Word order is thrown away, which sounds crude and works surprisingly well. scikit-learn's `CountVectorizer` does all of this; you'll write it yourself first, then use it.

**Multinomial Naive Bayes** is the classic model for word counts. For each class, it learns how likely each word is (`P(word | spam)`), and scores a message by adding up the logarithms of those probabilities for its words, plus the log of how common the class is. It's "naive" because it treats words as independent of each other, which they aren't; it's fast, needs little data, and is easy to explain, because the ratio `log P(word | spam) − log P(word | ham)` says how much each word pushes a message towards spam.

Write these functions:

- `load_messages(path)`: the messages, with a `spam` column (1 for spam, 0 for ham).
- `tokenise(text)`: the list of tokens, `re.findall(r"[a-z0-9£]+", text.lower())`.
- `vocabulary(texts, min_df=1)`: the sorted list of tokens appearing in at least `min_df` of the messages (count each message once per token).
- `vectorise(texts, vocab)`: a NumPy array of shape `(messages, len(vocab))` of whole-number counts; tokens not in `vocab` are ignored.
- `spam_model()`: an unfitted pipeline of `CountVectorizer(tokenizer=tokenise, token_pattern=None, lowercase=False)` and `MultinomialNB()`.
- `evaluate(df, seed=0)`: split 75/25, stratified by `spam`, with `random_state=seed`; fit a `spam_model()` on the training texts; return `(model, report)` where the report has `accuracy` and `baseline` (always predicting the training majority class) on the test set, `confusion` (`tp`, `fp`, `fn`, `tn`, spam positive), `precision` and `recall` (all 3 places), and the test texts it got wrong: `missed` (spam let through) and `false_alarms` (ham flagged), in test order.
- `spammiest(model, n)`: the `n` vocabulary words with the largest `log P(word | spam) − log P(word | ham)` (from the fitted Naive Bayes step's `feature_log_prob_`), as `(word, value)` pairs with values rounded to 3 places; ties keep alphabetical order.
- `classify(model, texts)`: a list of `"spam"` or `"ham"`.

## Forge bonus: Naive Bayes from scratch

Multinomial Naive Bayes is only a few lines of NumPy. For each class `c`, count how often each word appears across that class's training messages, add a **smoothing** constant `α` to every count (so a word never seen in spam doesn't get probability 0, which would make one word veto a whole message), and divide by the class's total: that's `P(word | c)`. Score a message by `log P(c) + Σ count(word) × log P(word | c)` and pick the highest-scoring class. Turning scores into probabilities needs the **log-sum-exp** trick: subtract the largest score before exponentiating, so nothing overflows.
