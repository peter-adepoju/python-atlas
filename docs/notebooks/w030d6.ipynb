```python
class Box:
    def say(self, t):
        return f"[{t}]"

class Loud:
    def __init__(self, inner):
        self.inner = inner
    def say(self, t):
        return self.inner.say(t.upper())

class Twice:
    def __init__(self, inner):
        self.inner = inner
    def say(self, t):
        return self.inner.say(t + t)

print(Loud(Twice(Box())).say("hi"), Twice(Loud(Box())).say("hi"))
```

## Forge: a bike-hire scheme

A city runs a hire scheme with ordinary bikes, electric bikes and scooters. Riders hire a vehicle by its code and pay when they return it, by the minute. Members pay less, and on some days a price cap applies. Here's the specification. Amounts are whole numbers of pence.

**Exceptions**

- `HireError(Exception)`: the family.
- `UnknownVehicle(HireError)`, made as `UnknownVehicle(code)`, with the message `no vehicle X9` and the attribute `code`.
- `NotAvailable(HireError)`, made as `NotAvailable(code)`, with the message `E1 is not available` and the attribute `code`.

**Vehicles**

- `Vehicle(code)` with class attributes `UNLOCK = 100` and `PER_MIN = 10`, and an attribute `out` (False when made: whether it's on hire). `cost(minutes)` returns `UNLOCK + PER_MIN * minutes`; `minutes` must be an `int` of 0 or more (not a `bool`), otherwise `ValueError`. `can_hire()` returns True when it isn't out. `repr` gives `Bike('B1')`, with the real class name.
- `Bike(Vehicle)`: `PER_MIN = 5`.
- `BatteryMixin`: its `__init__(self, code, battery=100)` passes `code` on with `super().__init__(code)` and stores `battery` (an `int` from 0 to 100, otherwise `ValueError`). It extends `can_hire()` with `super()`: also needs a battery of at least 20. `use(minutes)` takes `minutes` off the battery, stopping at 0.
- `EBike(BatteryMixin, Bike)`: `PER_MIN = 12`.
- `Scooter(BatteryMixin, Vehicle)`: `UNLOCK = 50`, `PER_MIN = 15`.

No vehicle class rewrites `cost`; the battery classes don't rewrite `__repr__`.

**Pricing rules** (each with `price(vehicle, minutes)`)

- `Standard`: the vehicle's own `cost(minutes)`.
- `Member`: the vehicle's cost without the unlock fee.
- `DayCap(inner, cap)`: wraps another rule; its price, but never more than `cap`.

**`Scheme(pricing=None)`**: holds its vehicles, a `pricing` rule (a `Standard()` if none is given; it can be changed at any time), and a list `trips`.

- `add(vehicle)`: a second vehicle with the same code raises `ValueError`.
- `available()`: the codes of every vehicle that `can_hire()`, sorted.
- `hire(code)`: raises `UnknownVehicle` for an unknown code and `NotAvailable` if the vehicle can't be hired; otherwise marks it out.
- `return_vehicle(code, minutes)`: raises `UnknownVehicle` for an unknown code and a plain `HireError` (message `B1 is not out`) if it isn't out. Otherwise works out the price with the pricing rule, then, if the vehicle has a `use` method, calls it with the minutes, marks the vehicle back in, records `(code, minutes, price)` in `trips`, and returns the price. Bad `minutes` raise `ValueError` and change nothing.
- `takings()`: the total of all trip prices.
