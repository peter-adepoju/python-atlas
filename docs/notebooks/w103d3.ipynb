```python
# Run this cell once. It lets you check exercise answers inside this notebook.
# Workflow: run your exercise cell, then immediately run the "check" cell below it.
import io, builtins

def check(tests_src):
    """Run an exercise's tests against the cell you ran just before this one."""
    ip = get_ipython()
    code = ip.user_ns["In"][-2]
    def run(feed=(), echo=False):
        out = io.StringIO(); q = [str(x) for x in feed]; real_in, real_print = builtins.input, builtins.print
        def fake_input(p=""):
            if not q:
                raise AssertionError("The program asked for more input than the test gave it.")
            v = q.pop(0); out.write(str(p) + (v + "\n" if echo else "")); return v
        def fake_print(*a, **k):
            k.setdefault("file", out); real_print(*a, **k)
        builtins.input, builtins.print = fake_input, fake_print
        try:
            exec(code, {"__name__": "__main__"})
        finally:
            builtins.input, builtins.print = real_in, real_print
        return out.getvalue()
    norm = lambda t: "\n".join(l.rstrip() for l in str(t).rstrip().splitlines())
    def expect_output(want, inputs=(), echo=False):
        got = run(inputs, echo)
        assert norm(got) == norm(want), f"expected {norm(want)!r} but got {norm(got)!r}"
    import re
    uses = lambda *names: all(re.search(r"\b" + re.escape(n) + r"\b", code) for n in names)
    def run_with(subs, feed=(), echo=False):
        nonlocal code
        saved = code
        for old, new in subs.items():
            code = code.replace(old, new, 1)
        try:
            return run(feed, echo)
        finally:
            code = saved
    env = dict(ip.user_ns, run=run, norm=norm, source=code, expect_output=expect_output, uses=uses, run_with=run_with)
    try:
        env["output"] = run()
    except AssertionError:
        env["output"] = ""
    exec(tests_src, env)
    tests = [(k, v) for k, v in env.items() if k.startswith("test_") and callable(v)]
    ok = 0
    for name, fn in tests:
        label = (fn.__doc__ or name[5:].replace("_", " ")).strip()
        try:
            fn(); ok += 1; print("✓", label)
        except AssertionError as e:
            print("✗", label + ":", e or "not quite yet")
    print(f"{ok} of {len(tests)} checks passed.")
```

## The exam, looked back on

Score the exam honestly before you read on: part 1 out of 105, part 2 out of 75. **108** of 180 (60%) is a pass and **144** (80%) a distinction. Then look at *where* the marks went, not just how many:

- **Quiz questions you missed** point to a phase to reread: the questions came in course order, so the numbers tell you which.
- **Problems you didn't finish** are usually a time problem rather than a knowledge one. Did you stay too long on one? The budget was about 20 minutes a problem.
- **Checks that failed** in Part C or Part D are the most useful of all: each names one thing the specification asked for that your program didn't do. Read the specification again, line by line, against your code. Most lost marks in a larger program come from a rule read too quickly (an order of checks, a boundary such as "more than" or "at least"), not from not knowing how.

Write down three things to revisit, and when. The exam's problems stay on their pages: in a month, try the ones you missed again, cold.

## What a portfolio is for

A portfolio is read by someone with little time: someone deciding whether to work with you, a project maintainer, or a fellow learner. They'll spend about a minute on each project. In that minute they want to know:

- **What it does**, in one sentence they don't need to decode.
- **That it works**: a demo they can try, or a few commands to run it.
- **That it's built with care**: tests, a README with install, usage and test instructions, and a licence.
- **What you did**, honestly: your own project, your part in a team's, or your merged contribution.

Three or four strong projects beat ten half-finished ones. Choose them to show **range** (one with algorithms, one with the web or databases, one with data) rather than three of the same kind.

## Checking each project

Treat each project like code under review, with a checklist:

1. **A summary** of at most 100 characters, so it fits one line in a listing.
2. **A README** with Install, Usage, Tests and Licence sections.
3. **Tests**, and they pass on a fresh copy.
4. **A licence** file (MIT is a common, permissive choice).
5. **A demo** link or a recording, if the project has anything to see.

A small script turns this into a list of fixes per project, which is easier to act on than a vague sense that "it needs tidying".

## Choosing and describing

To choose, score each project (more for tests, a demo and finished checklist items) and then pick greedily: take the best, then prefer projects that add tags you haven't shown yet. To describe, use a fixed shape so the listing reads evenly:

> **Waymark**: a walking-route planner: shortest routes, times with climbing, round walks and a walk log. Python, 2,140 lines, 186 tests.

Be truthful. Say "built in a course" for course projects; don't claim users you don't have; credit anyone whose work you used, as this course credits its sources.
