### Bug hunt: Depth first is not fewest steps

Depth-first search finds **a** route, not the shortest, and here it wanders round the grid and prints `6`. Use breadth-first search so it prints `2`.

```python
grid = ["...", "...", "..."]

def steps(start, end):
    seen, stack = {start: 0}, [start]
    while stack:
        r, c = stack.pop()
        for nr, nc in ((r - 1, c), (r, c - 1), (r, c + 1), (r + 1, c)):
            if 0 <= nr < 3 and 0 <= nc < 3 and (nr, nc) not in seen:
                seen[(nr, nc)] = seen[(r, c)] + 1
                stack.append((nr, nc))
    return seen[end]

print(steps((0, 0), (0, 2)))
```

### Bug hunt: A network with a loop in it

This adds every short path without checking whether its ends are already joined, so it pays for a loop and prints `5.5` for three junctions where `3.5` is enough. Skip paths that join two junctions already in the same group so it prints `3.5`.

```python
paths = [("a", "b", 1.5), ("b", "c", 2.0), ("a", "c", 2.0)]
total, joined = 0, set()
for a, b, km in sorted(paths, key=lambda p: p[2]):
    total += km
    joined |= {a, b}
print(total)
```

## Warm-up

Breadth-first search reaches every cell in order of distance, so the first time it reaches the goal is by a shortest route. Mark cells as seen when you **add** them to the queue, not when you take them off, or the same cell is queued many times.

## Standard moves

- **Fewest steps** (unweighted): BFS with a queue and a `dist` dictionary.
- **Cycle in a directed graph** (circular dependencies): depth-first search with three colours: white (unvisited), grey (on the current path), black (finished). Reaching a grey node is a cycle.
- **Two sides** (teams, schedules, "no two neighbours alike"): colour by BFS, alternating; an edge between equal colours means it's impossible.
- **Shortest weighted route**: Dijkstra with a heap, remembering each node's **previous** node, then walking back from the end.
- **Cheapest network joining everything**: Kruskal's algorithm (Joseph Kruskal, 1956): sort edges by cost and add each one that joins two different groups (union-find from last week).
- **Longest chain of dependent tasks**: the critical path: process tasks in topological order, each finishing at its duration plus the latest finish among its dependencies.
