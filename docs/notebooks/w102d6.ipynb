## When you're stuck

You will be stuck at some point; everyone is. Have a routine ready, the one you've practised all course:

1. **Reread the question**, slowly. Half of "stuck" is a misreading.
2. **Work a tiny example by hand** and write down what the answer should be.
3. **Write the simple version** first, even if it's slow: a correct slow answer scores; an unfinished clever one doesn't.
4. **Print** inside your code to see what's really happening.
5. **Move on** if it's still not working after your budget, and come back with fresh eyes.
