> **Why mod 97 catches typos.** Changing one base-32 character changes the number by d × 32ⁱ, where d is between −31 and 31 (not 0). For the check to miss it, 97 would have to divide that change. But 97 is prime, larger than 31, and shares no factor with 32, so it never divides d × 32ⁱ: every single-character error is caught. The same reasoning, with a prime bigger than the alphabet, is why ISBNs use 11 and bank numbers use 97.

## Forge: a ticket office

A theatre sells tickets whose codes are printed and sometimes read out over the phone, so they must be short, hard to mistype, and checked for errors. Each ticket has an **event** (0 to 1023, 10 bits), a **day** of the season (0 to 511, 9 bits) and a **seat** (0 to 4095, 12 bits). Write `TicketOffice()` (no arguments) with:

- `pack(event, day, seat)`: one 31-bit number, `event << 21 | day << 12 | seat`; `ValueError` if a field doesn't fit. `unpack(value)` reverses it.
- `check_digits(value)`: the two check digits of **ISO 7064 MOD 97-10** (used in IBAN bank account numbers): `98 - (value * 100) % 97`, so that `(value * 100 + check) % 97 == 1`. Any single wrong digit, and most swaps, then fail the check.
- `code(event, day, seat)`: `value * 100 + check`, written as exactly 8 characters in Douglas Crockford's base 32 alphabet, `0123456789ABCDEFGHJKMNPQRSTVWXYZ` (no I, L, O or U, which are easily confused or rude).
- `read(code)`: the `(event, day, seat)` back. Accept lower case, ignore hyphens, read `I` and `L` as `1` and `O` as `0`, and raise `ValueError` for any other character not in the alphabet or if the check fails.
- `draw(codes, k, seed)`: k different winning codes, chosen fairly and repeatably with a **partial** Fisher-Yates shuffle (for i from 0 to k − 1, swap position i with a random position from i to the end, using `random.Random(seed).randint`), in the order drawn. `ValueError` if k is impossible.
- `chance_any(n, k, m)`: the exact probability, as a `Fraction`, that at least one of m tickets you hold is among k winners drawn from n tickets: 1 − C(n − m, k) / C(n, k).
