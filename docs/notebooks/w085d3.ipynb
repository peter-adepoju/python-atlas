```python
import bisect
import zlib

def h(text):
    return zlib.crc32(text.encode())

def ring(nodes, vnodes=50):
    points = sorted((h(f"{n}#{i}"), n) for n in nodes for i in range(vnodes))
    return [p for p, _ in points], [n for _, n in points]

def owner(key, r):
    hashes, names = r
    return names[bisect.bisect(hashes, h(key)) % len(hashes)]

keys = [f"user{i}" for i in range(1000)]
before = ring(["a", "b", "c"])
after = ring(["a", "b", "c", "d"])
moved = sum(owner(k, before) != owner(k, after) for k in keys)
print(moved, "of 1000 keys moved")
print(sum(owner(k, after) == "d" for k in keys) == moved)
```

```python
import random

def backoff_delays(attempts, base, cap, rng):
    return [round(rng.uniform(0, min(cap, base * 2 ** i)), 3) for i in range(attempts)]

print(backoff_delays(5, 0.1, 2.0, random.Random(1)))
slo, total, failed = 0.999, 2_000_000, 1300
budget = (1 - slo) * total
print(round(budget), round(failed / budget, 2), "of the budget used")
print(round(0.999 * 0.999, 6), round(1 - 0.001 ** 2, 6))
```

```python
from sklearn.datasets import load_wine
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = load_wine(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0, stratify=y)
search = GridSearchCV(make_pipeline(StandardScaler(), KNeighborsClassifier()),
                      {"kneighborsclassifier__n_neighbors": [1, 3, 5, 9]}, cv=5).fit(X_train, y_train)
baseline = DummyClassifier(strategy="most_frequent").fit(X_train, y_train).score(X_test, y_test)
print(search.best_params_, round(search.best_score_, 3))
print("test:", round(search.score(X_test, y_test), 3), "baseline:", round(baseline, 3))
```

### Bug hunt: A percentile off by one

## System design

- **Estimate first** (Week 81): requests per second from daily users and actions (a day is about 10⁵ seconds), peak as a multiple of average, storage per year, and servers at a safe utilisation (about 70%). **Little's law**: requests in progress = arrival rate × time in the system.
- **Caches** (LRU, with a time to live) take load off slow stores; the hard part is **invalidation** (delete or update the cached copy when the data changes) and **stampedes** (many misses at once).
- **Queues** decouple producers from consumers, smooth bursts and allow retries; consumers must be **idempotent** because messages can arrive twice.
- **Replicas** spread reads and survive failures (with replication lag); **partitions** (shards) spread writes; **consistent hashing** moves only about 1/n of the keys when a node joins.
- **Distributed systems** (Week 82): no shared clock (Lamport and vector clocks order events), majorities (quorums, Raft) agree on one history, and sagas undo multi-step work with compensations.

Every key that moved went to the new node, and only about a quarter moved. With `hash(key) % n`, most keys would have moved.

## Reliability

- **Measure what users feel**: rate, errors and duration (RED); percentiles, not averages; traces that follow one request through many services (Week 82).
- **Service level objectives**: an SLO such as "99.9% of requests succeed" leaves an **error budget** of 0.1%; the **burn rate** is how fast you're spending it (1 means exactly on budget). Alert on fast burns over short and long windows together.
- **Retries** only for transient failures and idempotent operations, with **exponential backoff** and **jitter** (so clients don't retry in step), a cap, and a retry budget. **Circuit breakers** stop calling a failing dependency and try again after a pause. **Deadlines** bound the whole request, not each step.
- **Availability** multiplies in series (0.999 × 0.999 ≈ 0.998) and improves in parallel (1 − 0.001²).

## Data

- **NumPy** (Week 83): whole-array operations, broadcasting, boolean masks, `axis` and `keepdims`, `nan` with `np.isnan` and the `nan`-functions, seeded `default_rng`.
- **pandas**: `read_csv` (with `dtype=str` for messy files), `loc` and `iloc`, masks with `&` and `|`, new columns with `assign`, `groupby().agg(name=(column, function))`, `transform`, `merge(validate=..., indicator=True)`, `pivot_table`, `to_datetime` and `resample`. Change rows with one `.loc`, never chained indexing.
- **Cleaning**: profile first, never edit the raw file, parse each column deliberately, count and report what you drop.
- **Charts** (matplotlib): line for change over time, bar for categories (sorted, from zero), scatter for relationships, histogram for distributions; label everything, with units.

## Machine learning

- **The workflow** (Week 84): split off a test set first (stratified for classes); build a **pipeline** (scaling, encoding, model) so preprocessing is fitted on training data only; tune hyperparameters with **cross-validation** on the training set (`GridSearchCV`); score on the test set **once**; compare with a **baseline** (`DummyClassifier`).
- **Metrics**: accuracy misleads on imbalanced classes; use the confusion matrix, **precision** (flags that were right) and **recall** (positives caught), choose the threshold by the costs of each mistake, and ROC AUC to compare rankings; MAE, RMSE and R² for regression.
- **Models**: linear and logistic regression (explainable), nearest neighbours (needs scaling), trees and forests (flexible, can overfit), Naive Bayes (text), neural networks (layers, activations, backpropagation, gradient descent); training error far below test error means **overfitting**.

## Classic mistakes

- Averages instead of percentiles; caches with no invalidation; `hash(key) % n` sharding; retries on non-idempotent requests, or without jitter.
- Chained indexing in pandas; joins that duplicate rows; filling missing values with zero before averaging; bars that don't start at zero.
- Scaling before splitting; tuning on the test set; reporting accuracy on imbalanced data; judging a model by its training score.
