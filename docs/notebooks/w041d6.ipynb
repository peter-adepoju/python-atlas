> **Paying in advance.** Look at where `SensorLog` spends its time. Each `add` does a little extra work (one addition for the prefix sums, and occasionally a resize); in return, every `total` is instant, and `best_window` and `count_totals` never add up a range. `calibrate` is the one operation that disturbs the summary, so it rebuilds it once, at a cost of O(n), rather than patching it after every correction. That's the trade at the heart of data structure design: decide which operations must be fast, and pay for them, in advance or in memory, out of the operations that can afford it.

## Forge: a sensor log

A weather station records one whole-number reading a minute. Volunteers ask questions about the log, and occasionally correct whole stretches of it after finding that a sensor was off. Build `SensorLog`, which answers every question quickly, even with hundreds of thousands of readings.

**Storage.** Readings go into your own dynamic array: a fixed block made with `[None] * capacity`, starting at capacity 1 and doubling when full, never appended to. Keep `resizes` (how many times the block grew) and a read-only property `capacity`. Keep **prefix sums** up to date as readings arrive, so range totals never need a loop.

- `add(value)` stores a reading; `len(log)` is how many there are; `log[i]` reads one (negative indexes count from the end; out of range raises `IndexError`).
- `total(i, j)`: the sum of readings `i` to `j - 1`, in O(1). `i` and `j` must satisfy `0 <= i <= j <= len`, else `IndexError`.

**Questions.**

- `best_window(k)`: the busiest `k` consecutive minutes, as `(start, total)`; earliest on a tie. `ValueError` if `k` isn't from 1 to `len`.
- `longest_between(lo, hi)`: the longest run of consecutive readings all from `lo` to `hi` inclusive, as `(start, end)` (half-open, earliest on a tie), or `None` if no reading is in range.
- `count_totals(k)`: how many runs of consecutive readings (length 1 or more) add up to exactly `k`. Readings may be negative.
- `closest_pair(target)`: two **different readings** (at different positions) whose sum is closest to `target`, as `(a, b)` with `a <= b`; on a tie, the smaller sum. Use a sorted copy and two pointers. Fewer than 2 readings raises `ValueError`.

**Corrections.** `calibrate(corrections)`, where each correction is `(i, j, delta)`: add `delta` to every reading from `i` to `j - 1`. Apply them all at once with a **difference array** in O(n + number of corrections), then bring the prefix sums back up to date (once, not once per correction).

The checks compare every answer with a simple, slow version on random data, and time the big cases.
