## Warm-up

The session started well: an open question, and the learner found the cause themselves. Then the mentor took over with a leading question and a long explanation, and the learner said "Okay." That's the commonest way mentoring goes wrong, and you can see it in the numbers.

## Questions that help someone think

- **Open** questions ("what do you think happens to a month with no walks?") invite thinking, and are the mentor's main tool.
- **Closed** questions ("does it work?") get yes or no; useful for checking, not for thinking.
- **Leading** questions ("don't you think a defaultdict would be better?") are answers in disguise: the learner agrees without understanding.

When the learner is stuck, ask a smaller question rather than giving the answer. When they've found it, ask what they'd do next. Silence is fine: give them time.

## Pairing and listening

Pair with the **learner driving** (at the keyboard) and you **navigating** (thinking ahead, asking): their hands, their learning. In longer sessions swap every ten minutes or so, so they also see how you'd think. Listen for what they **don't** say: "Okay" after a long explanation usually means "I'm lost and too polite to say so". A useful rule: the learner should talk at least as much as you.

## Goals and notes

End each session with **one or two goals** for next time, small enough to finish: specific (what exactly), measurable (how you'll both know), and dated ("by Friday: the monthly summary shows every month, with a test for an empty February"). Keep short **notes**: what you talked about, what's done, what's in the way, the goals, so next week starts where this one stopped.
