## Predict, observe, explain

The sequence this course uses daily has a long history in teaching (often called **predict, observe, explain**): ask the learner to commit to a prediction, show what really happens, then explain the gap. Committing makes the learner care about the answer, and the surprise marks the memory. Then **check**: a second, slightly different question shows whether the idea transferred. A good five-minute explanation might be: a question (30 seconds), the reveal (30), the idea named and shown two ways (2 minutes), a check question (1 minute), and a summary (30).

## Misconceptions

Beginners don't arrive empty: they bring **models** of how things work, some wrong in predictable ways. "Assignment copies", "a function can't change my variable", "`0.1 + 0.2 == 0.3`". The best explanations are built on them: collect learners' actual wrong answers (from quizzes, questions, forums), count them, and aim the explanation at the most common. The same wrong models give you **plausible wrong options** for a prediction question, which is far better than random ones: each wrong option diagnoses a particular misunderstanding.
