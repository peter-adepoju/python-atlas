````python
def push(heap, value):
    heap.append(value)
    i = len(heap) - 1
    while i > 0:
        parent = (i - 1) // 2
        if heap[i] < heap[parent]:
            heap[i], heap[parent] = heap[parent], heap[i]
            i = parent
        else:
            break

heap = []
for v in [5, 3, 8, 1, 4, 2]:
    push(heap, v)
    print(f"push {v}: {heap}")
````

````python
def sift_down(heap, i):
    n = len(heap)
    while True:
        smallest = i
        for child in (2 * i + 1, 2 * i + 2):
            if child < n and heap[child] < heap[smallest]:
                smallest = child
        if smallest == i:
            return
        heap[i], heap[smallest] = heap[smallest], heap[i]
        i = smallest

def pop(heap):
    if not heap:
        raise IndexError("pop from an empty heap")
    last = heap.pop()
    if not heap:
        return last
    top, heap[0] = heap[0], last
    sift_down(heap, 0)
    return top

heap = [1, 3, 2, 7, 4, 5]
out = []
while heap:
    out.append(pop(heap))
    print(heap)
print(out)
````

````python
def sift_down(heap, i):
    n = len(heap)
    while True:
        smallest = i
        for child in (2 * i + 1, 2 * i + 2):
            if child < n and heap[child] < heap[smallest]:
                smallest = child
        if smallest == i:
            return
        heap[i], heap[smallest] = heap[smallest], heap[i]
        i = smallest

def heapify(a):
    for i in range(len(a) // 2 - 1, -1, -1):     # the last parent is at len // 2 - 1
        sift_down(a, i)

a = [9, 4, 7, 1, 8, 2, 6, 3, 5]
heapify(a)
print(a, all(a[(i - 1) // 2] <= a[i] for i in range(1, len(a))))
````

> **Key idea.** A binary min-heap is a complete binary tree stored in a list (children of i at 2i + 1 and 2i + 2, parent at (i − 1) // 2) where every node is no larger than its children, so the minimum is at index 0. Push appends and sifts up, swapping with the parent while smaller; pop moves the last item to the root and sifts down, swapping with the smaller child while larger. Both are O(log n), and the height is always ⌊log₂ n⌋. Heapify sifts down every parent from the last (index n // 2 − 1) to the root, in O(n). A max-heap flips the comparisons. Heaps are the standard priority queue.

### Bug hunt: The first smaller child

This sift-down swaps with the **first** child that's smaller, not the smallest, so a larger child can end up above a smaller one: this prints `[2, 9, 1]`, which isn't a heap. Swap with the smaller of the two children so it prints `[1, 2, 9]`.

````python
def sift_down(heap, i):
    n = len(heap)
    while True:
        for child in (2 * i + 1, 2 * i + 2):
            if child < n and heap[child] < heap[i]:
                heap[i], heap[child] = heap[child], heap[i]
                i = child
                break
        else:
            return

heap = [9, 2, 1]
sift_down(heap, 0)
print(heap)
````

## Warm-up

Every value is at least as big as its parent, so the smallest value must be at the top, at index 0. That's all a heap promises: it isn't sorted (3 comes before 2, and 7 before 4), and that looseness is exactly what makes it cheap to keep.

## The heap rules

A **binary min-heap** is a binary tree with two rules:

1. **Shape**: it's **complete**: every level is full except perhaps the last, which fills from the left. So it can live in a list with no gaps (Week 45 Day 2): the children of index `i` are at `2i + 1` and `2i + 2`, and its parent at `(i - 1) // 2`.
2. **Order**: every node is **less than or equal to** its children. So the minimum is always at index 0.

```text
             1                 index:  0  1  2  3  4  5
           /   \               value: [1, 3, 2, 7, 4, 5]
          3     2
         / \   /               3's children: 7 and 4 (both ≥ 3)
        7   4 5                2's child: 5 (≥ 2)
```

A **max-heap** flips the order rule (every node ≥ its children), keeping the maximum at the top. Unlike a BST, a heap says nothing about left versus right: 3 is on the left of 2 here, and it could just as well have been the other way round. That's why a heap can't search for an arbitrary value quickly (it's O(n)), but it doesn't need to: its job is the minimum.

The shape rule gives a height of ⌊log₂ n⌋, always, with no rotations: a heap of a million items is 19 levels deep.

## Push: sift up

To add a value, put it at the **end** of the list (the next free spot in the last level, so the shape stays complete). The order rule may now be broken between it and its parent, so **sift it up**: while it's smaller than its parent, swap them. It stops when its parent is no larger, or it reaches the root. At most one swap per level: O(log n).

## Pop: sift down

Taking the minimum is `heap[0]`, but removing it leaves a hole at the top. Fill it with the **last** item (popping the end of a list is O(1), and the shape stays complete), then **sift it down**: while it's larger than its smaller child, swap it with that **smaller** child. Swapping with the smaller one matters: it becomes the parent of the other child, so it must be the smaller of the two.

Popping everything gives the values in sorted order: that's heap sort, which is Day 3's topic.

## Heapify in O(n)

To make a heap from n values you could push them one at a time: O(n log n). There's a better way, published by Robert Floyd in 1964: treat the list as a complete tree already, and **sift down every parent, from the last one back to the root**. Leaves (the whole second half of the list) are already tiny heaps, so they need nothing; each parent, once sifted down, sits on top of two heaps and so makes a heap itself.

Why O(n)? Most nodes are near the bottom, where sifting down is short. Half the nodes are leaves (no work), a quarter sit one level up (at most 1 swap), an eighth two levels up (at most 2), and so on. The total is at most n × (1/4 + 2/8 + 3/16 + ...) = n × 1, so fewer than n swaps. Pushing one at a time is the other way round: the many items at the bottom are the ones that may travel the furthest.

## Priority queues

A **priority queue** is an abstract data type (like Week 43's stacks and queues): you add items with priorities, and you always take out the one with the **highest priority** next (here, the smallest number, as in "priority 1 first"). A heap is the usual way to build one:

| | add | look at the top | take the top |
|---|---|---|---|
| unsorted list | O(1) | O(n) | O(n) |
| sorted list | O(n) | O(1) | O(1) |
| binary heap | O(log n) | O(1) | O(log n) |
| balanced BST (Week 46) | O(log n) | O(log n) | O(log n) |

The balanced BST matches it, but the heap is simpler, uses a plain list with no node objects, and is much faster in practice. Priority queues run event simulations, schedulers, Huffman coding (Week 45), and next phase's shortest-path algorithms.

To store items with priorities, push **tuples** `(priority, item)`: tuples compare by their first element, then the second. Tomorrow covers what to do when two priorities tie and the items can't be compared.
