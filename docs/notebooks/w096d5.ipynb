```python
import re
from datetime import date

def log_table(log):
    lines = ["| Project | Kind | Pull request | Title | Status |", "|---|---|---|---|---|"]
    for c in log:
        lines.append(f"| {c['repo']} | {c['kind']} | #{c['pr']} | {c['title'].replace('|', chr(92) + '|')} | {c['status']} |")
    return "\n".join(lines)

def link_refs(text, repo):
    text = re.sub(r"\bgh-(\d+)\b", lambda m: f"[gh-{m.group(1)}](https://github.com/{repo}/issues/{m.group(1)})", text)
    return re.sub(r"(?<![\w\[])#(\d+)\b", lambda m: f"[#{m.group(1)}](https://github.com/{repo}/pull/{m.group(1)})", text)

WORDS = {"docs": "docs", "tests": "tests", "bug": "a bug fix"}

def cv_line(log):
    projects = {}
    for c in log:
        if c["kind"] != "review" and c["status"] == "merged":
            projects.setdefault(c["repo"], []).append(c["kind"])
    parts = []
    for repo, kinds in projects.items():
        distinct = list(dict.fromkeys(WORDS[k] for k in kinds))
        joined = distinct[0] if len(distinct) == 1 else ", ".join(distinct[:-1]) + " and " + distinct[-1]
        parts.append(f"{len(kinds)} merged pull request{'s' if len(kinds) != 1 else ''} to {repo} ({joined})")
    reviews = sum(1 for c in log if c["kind"] == "review")
    if reviews:
        parts.append(f"{reviews} review{'s' if reviews != 1 else ''}")
    return "; ".join(parts)

def contribution_stats(log, today):
    d = date.fromisoformat
    prs = [c for c in log if c["kind"] != "review"]
    days = sorted((d(c["merged"]) - d(c["opened"])).days for c in prs if c["status"] == "merged")
    n, mid = len(days), len(days) // 2
    median = None if not days else days[mid] if n % 2 else (days[mid - 1] + days[mid]) / 2
    waiting = [(d(today) - d(c["opened"])).days for c in prs if c["status"] == "open"]
    return {"median_days_to_merge": "-" if median is None else format(median, "g"), "longest_open_days": max(waiting, default=0),
            "open": len(waiting)}

def write_up(log, today, repo_links):
    ___
```

## Warm-up

Five entries: three merged pull requests, one still open, and a review. A review counts: it's real work for a real project, and many people never do one. `contrib_log.py` is an example log; yours will look different.

## Why write it up

Memory is generous to the past: in six months you'll remember the merge, not the three review rounds that taught you how CPython's tests are organised. A write-up records what actually happened, which is what makes it useful to others: "how I got my first pull request into CPython" posts are read by thousands of people about to try. And in a portfolio, a link to a merged pull request, with a sentence on what it fixed, is concrete evidence of exactly the skills this phase was about.

## Honest numbers

Numbers make a write-up concrete, and honest ones make it trustworthy: three merged pull requests, one open, one review; a median of nine days from opening to merge; one pull request needed three rounds of review. Don't inflate (a typo fix is a typo fix, and that's fine) and don't hide the open or closed ones: a pull request that was declined, and why, is often the most interesting part. Link everything, so readers can check: `gh-140101` as a link to the issue, `#140188` as a link to the pull request.

## A post worth reading

A useful structure: **what** you did (the table), **how** it went (one paragraph per contribution: the problem, the fix, what review taught you), **what you'd tell a newcomer** (the specific, practical things: "build CPython before choosing an issue", "reviewers asked for subTest"), and **what next**. Specifics beat generalities; the mistakes are the most useful part. Publish it where people look (your own site, a blog, the project's discussion forum if it welcomes such posts) and link it from your portfolio page.
