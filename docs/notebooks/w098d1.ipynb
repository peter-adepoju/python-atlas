### Bug hunt: A list that lost its tail

This overwrites `current.next` before saving it, so the walk stops after one node and prints `[1]`. Save the next node first, so it prints `[3, 2, 1]`.

```python
from nodes import make_list, to_list

def reverse(head):
    previous, current = None, head
    while current is not None:
        current.next = previous
        previous = current
        current = current.next
    return previous

print(to_list(reverse(make_list([1, 2, 3]))))
```

### Bug hunt: Checking only the children

Each node here is larger than its left child and smaller than its right one, but 6 sits in 5's left subtree, so this isn't a search tree; checking only children prints `True`. Check bounds for whole subtrees so it prints `False`.

```python
from nodes import make_tree

def is_bst(node):
    if node is None:
        return True
    if node.left and node.left.value >= node.value:
        return False
    if node.right and node.right.value <= node.value:
        return False
    return is_bst(node.left) and is_bst(node.right)

print(is_bst(make_tree([5, 3, 8, 1, 6])))
```

## Warm-up

`make_tree` reads a tree in **level order**, with `None` for a missing child: 5 at the root, 3 and 8 below it, then 3's children (none on the left, 4 on the right). That's how interview problems usually write trees, so it's worth being able to read it at a glance.

## Pointers and trees

**Linked lists**: draw the boxes and arrows before coding, and name every reference you need (`previous`, `current`, `following`). Most bugs are a lost `next` because it was overwritten before it was saved. Two pointers moving at different speeds (Robert Floyd's "tortoise and hare") find the middle of a list or a cycle in O(1) extra space.

**Trees**: depth-first (recursive, or a stack) for questions about paths and depths; breadth-first with a queue for questions about levels. A **binary search tree** keeps every value in the left subtree smaller and every value in the right larger, which is a property of the whole subtree, not just a node's children: pass the allowed **bounds** down as you go.

Recursion depth: a tree of 100,000 nodes in a line is 100,000 calls deep. Say so in the interview, and use an explicit stack when it matters.
