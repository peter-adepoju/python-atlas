````python
# Run this cell once. It lets you check exercise answers inside this notebook.
# Workflow: run your exercise cell, then immediately run the "check" cell below it.
import io, builtins

def check(tests_src):
    """Run an exercise's tests against the cell you ran just before this one."""
    ip = get_ipython()
    code = ip.user_ns["In"][-2]
    def run(feed=(), echo=False):
        out = io.StringIO(); q = [str(x) for x in feed]; real_in, real_print = builtins.input, builtins.print
        def fake_input(p=""):
            if not q:
                raise AssertionError("The program asked for more input than the test gave it.")
            v = q.pop(0); out.write(str(p) + (v + "\n" if echo else "")); return v
        def fake_print(*a, **k):
            k.setdefault("file", out); real_print(*a, **k)
        builtins.input, builtins.print = fake_input, fake_print
        try:
            exec(code, {"__name__": "__main__"})
        finally:
            builtins.input, builtins.print = real_in, real_print
        return out.getvalue()
    norm = lambda t: "\n".join(l.rstrip() for l in str(t).rstrip().splitlines())
    def expect_output(want, inputs=(), echo=False):
        got = run(inputs, echo)
        assert norm(got) == norm(want), f"expected {norm(want)!r} but got {norm(got)!r}"
    import re
    uses = lambda *names: all(re.search(r"\b" + re.escape(n) + r"\b", code) for n in names)
    def run_with(subs, feed=(), echo=False):
        nonlocal code
        saved = code
        for old, new in subs.items():
            code = code.replace(old, new, 1)
        try:
            return run(feed, echo)
        finally:
            code = saved
    env = dict(ip.user_ns, run=run, norm=norm, source=code, expect_output=expect_output, uses=uses, run_with=run_with)
    try:
        env["output"] = run()
    except AssertionError:
        env["output"] = ""
    exec(tests_src, env)
    tests = [(k, v) for k, v in env.items() if k.startswith("test_") and callable(v)]
    ok = 0
    for name, fn in tests:
        label = (fn.__doc__ or name[5:].replace("_", " ")).strip()
        try:
            fn(); ok += 1; print("✓", label)
        except AssertionError as e:
            print("✗", label + ":", e or "not quite yet")
    print(f"{ok} of {len(tests)} checks passed.")
````

### Exercise: Forge: a project starter

Write these functions, working with real files in the current folder (use the starter's `write` helper).

1. `create_project(name, packages=())`: create the layout above under a folder `name`, and return the sorted list of the paths created (relative, using `/`). File contents:
   - `.gitignore`: the lines `.venv/`, `__pycache__/`, `*.pyc`, `.env`, `dist/`.
   - `README.md`: `# NAME`, a blank line, `## Setup`, a blank line, then the four lines `python3 -m venv .venv`, `source .venv/bin/activate`, `pip install -r requirements.txt`, `python3 -m NAME.main` each indented by four spaces (a Markdown code block).
   - `pyproject.toml`: `[project]`, `name = "NAME"`, `version = "0.1.0"`, `requires-python = ">=3.10"`, a blank line, `[tool.ruff]`, `line-length = 88`.
   - `requirements.txt`: one package per line, sorted (empty if none).
   - `src/NAME/__init__.py`: a one-line docstring `"""NAME."""`.
   - `src/NAME/main.py`: a module docstring, a `main()` that prints `Hello from NAME`, and the `__main__` guard.
   - `tests/test_main.py`: `from NAME.main import main` and a `test_main()` function containing `main()`.
   Every file ends with a newline. Refuse to overwrite: if the folder already exists, raise `FileExistsError(name)`.
2. `check_project(name)`: return a sorted list of problems with an existing project folder: `"missing FILE"` for each of `README.md`, `.gitignore`, `requirements.txt`, `pyproject.toml` that's absent; `".gitignore does not ignore .venv"` if it exists without a `.venv/` line; `"no main.py with a __main__ guard"` unless some `main.py` under `src/` contains `__name__ == "__main__"`; and `"code without docstring: PATH"` for each `.py` file under `src/` whose module has no docstring (`ast.get_docstring`).

````python
import ast
import os

def write(path, text):
    """Write text to path, making folders first."""
    os.makedirs(os.path.dirname(path) or ".", exist_ok=True)
    with open(path, "w") as f:
        f.write(text)
````

## Forge: a project starter

A tidy Python project, ready for Git, GitHub and everything this week, looks like this:

```
weather/
├── .gitignore            keeps .venv, caches and secrets out of Git
├── README.md             what it is and how to run it
├── pyproject.toml        settings for tools (and, later, packaging)
├── requirements.txt      the packages it needs
├── src/
│   └── weather/
│       ├── __init__.py
│       └── main.py       def main() ... if __name__ == "__main__": main()
└── tests/
    └── test_main.py      (Week 24)
```

Putting the code in `src/` keeps it separate from the project's other files, a common layout for real projects.

## Forge bonus: are the requirements right?

Two common mistakes: importing a package that isn't in `requirements.txt` (the project works on your computer and fails everywhere else), and listing packages that are never used. A tool can check both, with one complication: some packages are imported under a different name from the one you install.
