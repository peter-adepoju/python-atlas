The index does real work in `add` (counting every word, updating two maps) so that `search` touches only the documents that contain the query's rarest word, not every document. ### Exercise: Forge bonus: a Bloom filter

A **Bloom filter** (Burton Bloom, 1970) answers "have I seen this?" using far less memory than a set, by allowing one kind of mistake. It's a list of `m` bits, all 0, and `k` different hash functions. To **add** an item, set the `k` bits its hashes point to. To **check** an item, look at its `k` bits: if any is 0, the item was **definitely never added**; if all are 1, it was **probably** added (or other items happen to have set those bits: a false positive). There are never false negatives. Browsers and databases use them to skip expensive look-ups for things that certainly aren't there.

## Forge: a search engine

Search engines don't read every document for every query. They build an **inverted index** in advance: for each word, which documents contain it, and how often. A query then only looks up its own words. Build `SearchIndex` on a hash map of your own.

**Your hash map.** Write `HashMap`, with chaining (Day 2): `m[key]` (`KeyError` if missing), `m[key] = value`, `del m[key]`, `in`, `len`, `get(key, default=None)`, `keys()` and `items()` (lists, in any order), doubling when the load factor passes 0.75. **No dict or set anywhere in the Forge** (no `{}` displays, `dict()`, `set()` or `Counter`): every map below is your `HashMap`.

**Words.** `words(text)`: the lower-cased runs of letters and digits in `text`, in order (anything else separates words): `"Don't stop, 2 go!"` gives `["don", "t", "stop", "2", "go"]`.

**The index.** `SearchIndex()` keeps two `HashMap`s: `postings` from each word to a `HashMap` of document id to the number of times the word appears in that document; and `documents` from each document id to a `HashMap` of its words and their counts.

- `add(doc_id, text)`: index a document. Adding an id that's already there raises `ValueError`.
- `remove(doc_id)`: remove it from both maps (a word left with no documents is removed from `postings` entirely); a missing id raises `KeyError`.
- `search(query)`: the ids of documents containing **every** word of the query, ranked by the total number of times the query's words appear in them (most first), then by id. An empty query gives `[]`. Look up the query's **rarest** word first and check only its documents.
- `top_words(k)`: the `k` words that appear most often across all documents, as `(word, total)` pairs, most first, then alphabetically.
- `similar(doc_id)`: the other document sharing the most **distinct** words with it, or `None` if no other document shares any; ties go to the smallest id. Use the index: only documents in some of its words' postings can be candidates.
