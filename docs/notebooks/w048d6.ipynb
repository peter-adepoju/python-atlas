````python
def prefix_function(s):
    pi, k = [0] * len(s), 0
    for i in range(1, len(s)):
        while k > 0 and s[i] != s[k]:
            k = pi[k - 1]
        if s[i] == s[k]:
            k += 1
        pi[i] = k
    return pi

print(prefix_function("abcabcab"), len("abcabcab") - prefix_function("abcabcab")[-1])
````

> **Two ideas, one structure.** Aho-Corasick is what you get when two of this week's ideas meet. The trie shares work between patterns that **start** the same way; the failure links share work between positions in the text, never re-reading a character, as KMP did for a single pattern. Building the links breadth-first is the same "use what you've already computed" move as the prefix function's `k = pi[k - 1]`, applied to a tree instead of a line. The algorithm is still at the heart of tools that scan text for many fixed strings at once: the original `fgrep` command was built on it, and network intrusion detection and virus scanners use descendants of it to check traffic against thousands of signatures.

## Forge: many patterns at once

A message filter has a list of a thousand banned phrases, and must check every message for all of them. Running KMP once per phrase reads each message a thousand times. In 1975 Alfred Aho and Margaret Corasick, working on a bibliographic search tool, combined this week's two big ideas so the text is read **once**:

1. Put all the patterns in a **trie** (Day 1). Each node is "the last few characters read match this prefix of some pattern".
2. Give every node a **failure link** (Day 2's prefix function, generalised): the node for the **longest proper suffix** of its string that is also a prefix of some pattern (a path in the trie). On a mismatch, follow failure links until a node has the needed child, or the root is reached, exactly as KMP falls back with `pi`.
3. Compute the links **breadth-first** (Week 43): a node's failure link is found from its parent's, which is shallower and so already done. For a child reached from `node` by character `c`: start at `node.fail`, follow failure links until one has a child `c` (or you reach the root); that child (or the root, if even the root has none) is the link.
4. Give every node an **output list**: the patterns ending exactly there, followed by the output list of its failure link (patterns that end at the same place but are shorter, like "he" inside "she").

Then scanning the text is KMP with a trie: for each character, step (falling back as needed), and report everything in the current node's output list. The time is O(text length + total pattern length + number of matches).

```text
patterns: he, she, his, hers
          (root)
          /    \
         h      s
        / \      \
       e*  i      h
       |    \      \
       r     s*     e*     failure links: "sh" -> "h", "she" -> "he",
       |                    "hi" -> root, "his" -> "s", "her" -> root, ...
       s*                   output of "she": she, he
```

Write `AhoCorasick(patterns)`:

- Patterns must be non-empty strings (`ValueError` otherwise); repeats are ignored. Keep the distinct patterns in `patterns`, in first-seen order.
- `find_all(text)`: every match as `(start, pattern)`, sorted.
- `count(text)`: a dict from each pattern to how many times it occurs (overlaps count).
- `feed(ch)` and `reset()`: a streaming interface; `feed` returns the list of patterns that end with this character, longest first.
- `censor(text, mask="*")`: the text with every character that's part of **any** match replaced by `mask`. Do it in O(n): the longest match ending at each position covers all shorter ones, so mark its start and end in a difference array (Week 41's prefix sums) and sweep once.

No recursion. The checks compare with brute force on random patterns and texts, and time 1,000 patterns against a 100,000-character text.
