```python
class T:
    def __init__(self, v, l=None, r=None):
        self.v, self.l, self.r = v, l, r

def paths(n, so_far, out):
    if n is None:
        return out
    so_far = so_far + [n.v]
    if not n.l and not n.r:
        out.append(sum(so_far))
    paths(n.l, so_far, out)
    paths(n.r, so_far, out)
    return out

print(paths(T(1, T(2, T(3)), T(4)), [], []))
```

### Exercise: Forge bonus: Huffman coding

**Huffman coding** (David Huffman, 1952) compresses text by giving common characters short codes of bits and rare ones long codes, with no code the start of another, so the bits can be read back without separators. The codes come from a binary tree:

1. Count each character. Make a leaf for each, weighted by its count.
2. Repeatedly take the **two lightest** trees, join them under a new node whose weight is their sum (the lighter on the left), until one tree is left. ## Forge: an organisation chart

A company's staff list says who reports to whom, as text lines `name,manager` (the head of the company has an empty manager: `Ngozi,`). People ask the chart questions all day: who's in someone's team, who manages both of these people, how deep does the company go. Real data has mistakes, and the company is tall in places (a long chain of single reports), so **nothing may use recursion**: use explicit stacks and queues.

Build `OrgChart(lines)`:

- **Loading.** Each line is `name,manager` with spaces around each part ignored; blank lines are skipped. Raise `ValueError` for: a name listed twice, not exactly one head, a manager who isn't on the list, or a **cycle** (someone who, following managers upwards, never reaches the head). Keep each person's direct reports in the order their lines appear.
- `manager(name)` and `reports(name)` (direct reports, as a list); `chain(name)`: the names from `name` up to the head; unknown names raise `KeyError`.
- `team_size(name)`: everyone below `name` (not counting them).
- `common_manager(a, b)`: the lowest person who is `a` or above `a`, and also `b` or above `b`.
- `levels()`: a list of lists of names by depth (level order, reports in order).
- `outline()`: lines of the chart, depth-first (each person then their teams, in order), indented two spaces per level.
- `move(name, new_manager)`: reassign `name` (with their whole team) to report to `new_manager`, at the end of the new manager's reports. Moving the head, or moving someone under themselves or anyone in their own team, raises `ValueError` and changes nothing.
