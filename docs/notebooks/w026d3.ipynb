````python
# Run this cell once. It lets you check exercise answers inside this notebook.
# Workflow: run your exercise cell, then immediately run the "check" cell below it.
import io, builtins

def check(tests_src):
    """Run an exercise's tests against the cell you ran just before this one."""
    ip = get_ipython()
    code = ip.user_ns["In"][-2]
    def run(feed=(), echo=False):
        out = io.StringIO(); q = [str(x) for x in feed]; real_in, real_print = builtins.input, builtins.print
        def fake_input(p=""):
            if not q:
                raise AssertionError("The program asked for more input than the test gave it.")
            v = q.pop(0); out.write(str(p) + (v + "\n" if echo else "")); return v
        def fake_print(*a, **k):
            k.setdefault("file", out); real_print(*a, **k)
        builtins.input, builtins.print = fake_input, fake_print
        try:
            exec(code, {"__name__": "__main__"})
        finally:
            builtins.input, builtins.print = real_in, real_print
        return out.getvalue()
    norm = lambda t: "\n".join(l.rstrip() for l in str(t).rstrip().splitlines())
    def expect_output(want, inputs=(), echo=False):
        got = run(inputs, echo)
        assert norm(got) == norm(want), f"expected {norm(want)!r} but got {norm(got)!r}"
    import re
    uses = lambda *names: all(re.search(r"\b" + re.escape(n) + r"\b", code) for n in names)
    def run_with(subs, feed=(), echo=False):
        nonlocal code
        saved = code
        for old, new in subs.items():
            code = code.replace(old, new, 1)
        try:
            return run(feed, echo)
        finally:
            code = saved
    env = dict(ip.user_ns, run=run, norm=norm, source=code, expect_output=expect_output, uses=uses, run_with=run_with)
    try:
        env["output"] = run()
    except AssertionError:
        env["output"] = ""
    exec(tests_src, env)
    tests = [(k, v) for k, v in env.items() if k.startswith("test_") and callable(v)]
    ok = 0
    for name, fn in tests:
        label = (fn.__doc__ or name[5:].replace("_", " ")).strip()
        try:
            fn(); ok += 1; print("✓", label)
        except AssertionError as e:
            print("✗", label + ":", e or "not quite yet")
    print(f"{ok} of {len(tests)} checks passed.")
````

## Modules and the standard library

- `import x` runs the file `x.py` **once**, the first time, and keeps the module in `sys.modules`. Later imports reuse it.
- Python looks for modules in the folders on `sys.path`: first the script's own folder, then the standard library, then installed packages. A file of yours called `random.py` or `json.py` hides the real one.
- `if __name__ == "__main__":` runs code only when the file is run as a script, not when it's imported. Put a module's demo or command-line code under it.

| Module | Reach for it to... |
|---|---|
| `math` | square roots, `gcd`, `isqrt`, `inf`, `floor`/`ceil` |
| `random` | random choices; use `random.Random(seed)` for repeatable results |
| `statistics` | `mean`, `median`, `mode`, `stdev` |
| `datetime` | dates, times, differences (`timedelta`), formatting and parsing |
| `zoneinfo` | time zones: `ZoneInfo("Africa/Lagos")` |
| `collections` | `Counter`, `defaultdict`, `deque`, `namedtuple` |
| `re`, `json`, `csv` | patterns in text, and the two everyday data formats |
| `os`, `pathlib`, `sys` | files and folders, environment variables, command-line arguments |

## The command line and Git

The shell commands you'll use daily: `pwd`, `ls`, `cd`, `mkdir`, `cp`, `mv`, `rm`, `cat`, and `python3 script.py args`. Inside a script, `sys.argv` holds the arguments, and `argparse` turns them into named values with help and checking. A script's **exit status** (0 for success) and **standard error** let other programs and scripts know what happened.

The Git cycle:

```
git status                   # what's changed?
git add file.py              # stage a change
git commit -m "Add totals"   # save a snapshot
git log --oneline            # the history
git switch -c feature        # a new branch
git switch main
git merge feature            # bring the branch's commits in
git push                     # send commits to GitHub
git pull                     # fetch and merge others' commits
```

A **merge conflict** happens when two branches change the same lines; Git marks them with `<<<<<<<`, `=======` and `>>>>>>>`, you edit the file to what it should be, then `git add` and `git commit`. On GitHub, a **pull request** proposes a branch's changes for review before they're merged.

## Setting up a project

- **Install** a current Python, and check it with `python3 --version`.
- One **virtual environment** per project: `python3 -m venv .venv`, then activate it, then `pip install` what the project needs. `pip freeze > requirements.txt` records the versions; `pip install -r requirements.txt` recreates them elsewhere. Don't commit `.venv` itself.
- An editor with a **debugger** lets you pause at a breakpoint and inspect every variable, instead of adding prints.
- A **formatter** (such as Black) rewrites layout automatically; a **linter** (such as Ruff or Flake8) points out likely mistakes and style problems. PEP 8 is the style guide they follow.
- **Documentation**: `help(thing)` in Python, and the official docs at docs.python.org, which say exactly what each function takes, returns and raises.

## Data formats

The rules worth remembering: `fullmatch` to validate, `search` to find; JSON has no dates, sets or tuples, so convert them; always open CSV files with `newline=""`; store times in UTC and convert for display; never compare naive and aware datetimes.

## Testing

A test arranges inputs, acts by calling the code, and asserts the result. pytest finds `test_*.py` files and their `test_` functions, shows the values when an `assert` fails, and gives you `pytest.raises`, `pytest.approx`, `@pytest.mark.parametrize`, fixtures, and built-in fixtures such as `tmp_path`, `capsys` and `monkeypatch`. Good tests check boundaries and edge cases, are fast, independent and repeatable, test behaviour rather than implementation, and take expected values from somewhere other than the code.

## The classic mistakes

1. Naming your own file after a standard module (`random.py`, `csv.py`), so `import random` imports yours.
2. Demo code in a module without `if __name__ == "__main__":`, so it runs whenever the module is imported.
3. `re.match` or `re.search` where `fullmatch` was needed: `"12abc"` "matches" `\d+`.
4. A greedy `.*` that runs past the end of what you wanted: use `.*?` or a class like `[^"]*`.
5. `json.dump` on a set, a date or a custom object: `TypeError`.
6. Comparing a naive `datetime` with an aware one: `TypeError`; or doing arithmetic across a clock change in local time.
7. Tests that pass whatever the code does: no `assert`, or expected values computed by the same formula as the code.
8. Committing secrets, `.venv` or `__pycache__` to Git: list them in `.gitignore`.
