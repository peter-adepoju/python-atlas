### Bug hunt: A search that never ended

With `while lo < hi` the last candidate is never checked, so the search for `7` (the last item) misses it and prints `-1`. Fix the loop so it prints `3`.

```python
def search(nums, target):
    lo, hi = 0, len(nums) - 1
    while lo < hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            return mid
        if nums[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

print(search([1, 3, 5, 7], 7))
```

### Bug hunt: Greedy coins

Taking the biggest coin first gives `3` coins for 6 with `[1, 3, 4]`. Replace it with dynamic programming so it prints `2`.

```python
def fewest(coins, amount):
    count = 0
    for c in sorted(coins, reverse=True):
        count += amount // c
        amount %= c
    return count if amount == 0 else -1

print(fewest([1, 3, 4], 6))
```

## Warm-up

Taking the largest coin each time uses three coins (4, 1, 1), but two threes would do. Greedy choices that look right locally can be wrong globally; that's the signal for dynamic programming.

## Disguised structures

- "Sorted, but…" (rotated, with an unknown length, or a yes/no question that flips once): **binary search**, O(log n). Decide at each step which half is sorted, or which side the answer lies on.
- A grid of cells, a map, "connected", "reachable": a **graph**. Explore with **breadth-first search** and a queue (`collections.deque`), which has no recursion limit; a deep recursive search on a 300 by 300 grid can exceed Python's limit of about 1,000 calls.
- "Prerequisites", "must come before", "build order": a **directed graph**; a valid order is a **topological sort**, and a cycle means none exists.
- "Fewest", "how many ways", choices whose effects overlap: **dynamic programming**: the best answer for `amount` built from the best answers for smaller amounts.
- "The k-th largest so far", "the k closest": a **heap** of size k (`heapq`), O(log k) per item.
