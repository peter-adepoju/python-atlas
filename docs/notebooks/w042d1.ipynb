````python
class Node:
    def __init__(self, value, next=None):
        self.value, self.next = value, next

def from_list(values):
    head = None
    for v in reversed(values):
        head = Node(v, head)
    return head

def to_list(head):
    out = []
    node = head
    while node is not None:
        out.append(node.value)
        node = node.next
    return out

def length(head):
    count, node = 0, head
    while node is not None:
        count += 1
        node = node.next
    return count

def find(head, target):
    node = head
    while node is not None:
        if node.value == target:
            return node
        node = node.next
    return None

head = from_list([4, 8, 15, 16, 23, 42])
print(to_list(head), length(head), find(head, 16).value, find(head, 99))
````

````python
class Node:
    def __init__(self, value, next=None):
        self.value, self.next = value, next

class LinkedListIterator:
    def __init__(self, node):
        self._node = node

    def __iter__(self):
        return self

    def __next__(self):
        if self._node is None:
            raise StopIteration
        value = self._node.value
        self._node = self._node.next
        return value

class LinkedList:
    def __init__(self, values=()):
        self._head = self._tail = None
        self._size = 0
        for v in values:
            self.append(v)

    def push_front(self, value):
        self._head = Node(value, self._head)
        if self._tail is None:              # it was empty: the new node is also the tail
            self._tail = self._head
        self._size += 1

    def append(self, value):
        node = Node(value)
        if self._tail is None:
            self._head = self._tail = node
        else:
            self._tail.next = node
            self._tail = node
        self._size += 1

    def __len__(self):
        return self._size

    def __iter__(self):
        return LinkedListIterator(self._head)

    def __repr__(self):
        return f"LinkedList({list(self)!r})"

items = LinkedList([2, 3])
items.push_front(1)
items.append(4)
print(items, len(items), 3 in items, sum(items))
````

> **Key idea.** A linked list is a chain of nodes, each holding a value and a reference to the next, ending in `None`; the list itself is a reference to the head. Walk it with a separate variable, looping while `node is not None`. Adding or removing at the front is O(1), and so is splicing after a node you hold, but reaching item i is O(n). A `LinkedList` class keeps the head, the tail (for O(1) append) and the size (for O(1) `len`), and an iterator class makes it work with `for`, `in` and `list`. Every method that changes the list must keep the head, tail and size right.

## Warm-up

Three objects, each pointing to the next, and a loop that follows the pointers until it reaches `None`. That's a whole linked list: there's nothing else to it.

## Nodes and links

A **node** holds one value and a reference to the next node (`None` for the last one). The list as a whole is just a reference to the first node, the **head**:

```text
 head
  |
  v
+-----+---+    +-----+---+    +-----+------+
| ant | o-+--->| bee | o-+--->| cat | None |
+-----+---+    +-----+---+    +-----+------+
 value next     value next     value next
```

The nodes don't need to be next to each other in memory: each one could be anywhere, because the links say where to go next. That has two consequences, which run through this whole week:

- **Adding or removing a node only changes links.** To put a new node at the front, make it point to the old head and make it the new head: two assignments, O(1), however long the list is. Nothing shifts, unlike an array.
- **There's no arithmetic to find node i.** To reach it, you must start at the head and follow `i` links: O(n).

Pushing three words onto the front gives them in reverse order, the way a stack does.

## Walking a list

Almost every linked-list function has the same loop: start at the head, do something with each node, move to `node.next`, stop at `None`.

Two habits prevent most bugs:

1. **Loop on `node is not None`**, not `node.next is not None`, unless you really mean "stop at the last node" (which you sometimes do, to append).
2. **Use a separate variable** (`node`) to walk; moving `head` itself loses the start of the list.

## A LinkedList class

Passing bare head nodes around works, but a class can keep the head, the length (so `len` is O(1)), and a reference to the **tail** (so appending is O(1) too), and give the list Python's usual behaviour:

`__iter__` returns a fresh iterator each time (Week 36), so the list can be looped over again and again; and because the class is iterable, `list(...)`, `in`, `sum` and `for` all work without being written separately. The tail reference must be kept right by **every** method that changes the list: forgetting it in one place is the commonest linked-list bug.

## Arrays or links?

| Operation | Dynamic array (Python list) | Singly linked list (with head, tail and size) |
|---|---|---|
| read or write item i | O(1) | O(n): walk i links |
| add at the front | O(n): shift everything | **O(1)** |
| add at the back | O(1) amortised | O(1) |
| remove from the front | O(n) | **O(1)** |
| remove from the back | O(1) | O(n): must find the node before the tail |
| insert or delete after a node you're holding | O(n): shift | **O(1)**: relink |
| search | O(n) | O(n) |
| memory per item | one reference | a whole node object: value, link and object overhead |

A linked list wins when you add and remove at the front a lot, or splice at a node you already have; an array wins whenever you need to jump to positions. In Python in particular, a list is usually faster in practice even for operations with the same Big-O, because its references sit side by side in memory, which modern processors read very quickly, while nodes are scattered. When you need a fast queue, `collections.deque` (built from linked **blocks** of slots, a mixture of the two ideas) is the standard answer; Week 43 uses it.
