## Warm-up

Interviewers hear this answer every day, and it tells them only that the candidate prepared a dodge. A real weakness, with evidence of what you're doing about it, is far stronger, and today you'll build a check for the dodges.

## The hard questions

- **Conflict**: they want to see that you can disagree **respectfully**, understand the other view, and reach an outcome, not that you won. Describe the other person fairly ("they were worried about backports"), what you did to understand them, and what you agreed. Never blame.
- **Ambiguity**: they want to see you make progress without perfect information: what you assumed, how you checked the assumptions quickly (a prototype, a user test, a question to the right person), and how you adjusted.
- **Leading without authority**: influence by evidence and example: a measurement, a written proposal, offering to do the work.
- **Failure**: own it, say what you learned, and show the change in what you do now.

Your course has stories for all of these: the review disagreement in your first CPython pull request, the round-walk form that user testing showed was confusing, proposing to cut a feature with velocity numbers.

## Weaknesses and questions for them

**"What's your greatest weakness?"**: name a real one that won't disqualify you for the role, give a short example, and say what you're doing about it, with evidence ("I used to start coding before clarifying; since Week 97 I write the edge cases down first, and my last three mock rounds had no missed cases"). Avoid disguised strengths (perfectionism, working too hard).

**"Do you have any questions for us?"**: always have three or four, about things the website doesn't say: how the team decides what to build, what a new person's first month looks like, how code review works, what the interviewer enjoys about the work. They show interest and help you decide too.
