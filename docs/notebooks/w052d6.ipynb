> **Draft, then search.** The timetabler works in two stages, and real schedulers usually do the same. A greedy rule gives a decent answer almost instantly and tells you roughly how many slots you need. A backtracking search then finds a valid answer with fewer slots, or proves there isn't one, and it's only fast because of this week's pruning ideas: placing the most constrained exam first (as with sudoku cells), refusing slots that clash the moment they're tried, and treating slots as interchangeable so the same timetable with relabelled slots is never searched twice. With hundreds of exams even that gives out, and solvers turn to heuristics that improve a timetable step by step without promising the best.

## Forge: an exam timetabler

A school has to fit its exams into a few time slots so that no student sits two exams at once. Two exams **clash** if any student takes both; the clashes form a graph, and a timetable is a colouring of that graph with the slots as colours. Finding the fewest colours is a famously hard problem, so the timetabler drafts quickly with a greedy rule, then searches properly with backtracking.

Write `ExamTimetable(enrolments, slots)`, where `enrolments` maps each student to a list of exams and slots are numbered `0` to `slots - 1`:

- Raise `ValueError` if `slots` is less than 1. Store `slots`, and the clash graph as `graph`, a dict from every exam (anything a student takes) to the **set** of exams it clashes with. Unknown exams given to any method raise `KeyError`.
- `exams()` (sorted) and `clashes(exam)` (sorted).
- `fix(exam, slot)`: the exam must go in that slot (`ValueError` if the slot is out of range). Fixed exams are kept in `fixed`, a dict.
- `greedy()`: the Welsh-Powell draft: take exams in order of most clashes first (ties alphabetical), and give each the lowest slot no clashing exam already has. Return a dict; it may use more slots than allowed, and ignores fixed exams.
- `is_valid(timetable, slots=None)`: every exam is placed in a slot from 0 to slots − 1 (default `self.slots`), no clashing exams share a slot, and every fixed exam is where it was fixed.
- `schedule(slots=None)`: a valid timetable by backtracking, or `None`. Start from the fixed exams (return `None` if they clash). Each step, choose the unplaced exam whose clashing exams already use the **most different slots** (ties: more clashes, then alphabetical), and try its free slots in increasing order. When nothing is fixed, break symmetry: a new exam may use at most one slot beyond the highest slot used so far. Count every call of the search in `nodes`.
- `fewest_slots()`: the smallest number of slots a valid timetable needs (with the fixed exams; `0` with no exams).
- `back_to_back(timetable)`: the number of times a student has exams in two consecutive slots, summed over students (one count per pair of consecutive slots a student has).
