> **Key idea.** itertools gives lazy, C-speed stages. count, cycle and repeat are endless sources: bound them with zip, islice or takewhile. islice slices any iterator; takewhile and dropwhile act on the start of a stream; filterfalse and compress select. chain joins streams, zip_longest pads instead of truncating, pairwise gives neighbours, batched gives fixed-size groups, product gives nested-loop combinations. groupby groups neighbouring equal keys only, so sort first, and use each group before moving on. accumulate gives running results; permutations, combinations, product and combinations_with_replacement cover the four counting cases. tee stores whatever one copy is ahead by, so prefer a list when copies drift apart.

### Bug hunt: Groups that aren't together

`groupby` only groups **neighbouring** items, so unsorted words starting with the same letter make separate groups, each later group overwrites the earlier count, and this prints `{'a': 1, 'b': 1, 'c': 1}`. Sort by the same key first so it prints `{'a': 2, 'b': 2, 'c': 1}`.

```python
from itertools import groupby

words = ["apple", "bean", "avocado", "beet", "cherry"]
first = lambda w: w[0]
counts = {}
for k, g in groupby(words, key=first):
    counts[k] = len(list(g))
print(counts)
```

## Warm-up

## Endless sources

- `count(start=0, step=1)`: start, start + step, ... for ever (floats and negative steps work).
- `cycle(iterable)`: the items over and over (it keeps a copy of the first pass).
- `repeat(item, times=None)`: the same item for ever, or `times` times.

Pair an endless source with something that stops: `zip` against a finite iterable, `islice`, `takewhile`, or a `break`. `zip(count(1), names)` numbers names without `enumerate`; `zip(cycle(["red", "blue"]), rows)` alternates colours.

## Slicing and selecting

| Function | Yields |
|---|---|
| `islice(it, stop)` / `islice(it, start, stop, step)` | a slice of any iterator (no negative indexes) |
| `takewhile(pred, it)` | items while pred is True; stops at the first False |
| `dropwhile(pred, it)` | skips items while pred is True, then yields everything after |
| `filterfalse(pred, it)` | the items where pred is False (the opposite of `filter`) |
| `compress(data, selectors)` | the items of data whose selector is truthy |

`takewhile` and `dropwhile` look only at the **start**: once the condition fails, they stop checking. For "skip the header lines, then process the rest", `dropwhile` is exactly right.

## Combining

- `chain(a, b, ...)` and `chain.from_iterable(iterable_of_iterables)`: one stream after another; the second flattens one level, lazily.
- `zip_longest(a, b, fillvalue=None)`: like `zip`, but continues to the end of the **longest**, filling gaps. (`zip` stops at the shortest, silently; since Python 3.10, `zip(a, b, strict=True)` raises if the lengths differ.)
- `pairwise(it)`: each item with the next, `(a, b), (b, c), ...` (3.10).
- `batched(it, n)`: tuples of n items, the last possibly shorter (3.12), the same job as Monday's `Chunks`.
- `product(a, b, ...)`: every combination of one item from each, like nested loops; `product(range(6), repeat=2)` is two dice.

## Grouping

`groupby(iterable, key=None)` yields `(key, group)` pairs for each run of **neighbouring** items with the same key. It's run-length grouping, not SQL's GROUP BY: equal keys that aren't next to each other form separate groups, so **sort by the same key first** if you want one group per key. Each group is itself an iterator that shares the underlying stream: use (or copy) it **before** moving to the next group, because advancing `groupby` discards it.

## Accumulating and combinatorics

`accumulate(iterable, func=operator.add, initial=None)` yields running results: totals, products, running maxima, balances. `initial` puts a starting value in front.

The combinatoric four, all lazy and all in a fixed order (lexicographic by position):

| | Order matters | Order doesn't |
|---|---|---|
| no repeats | `permutations(it, r)` | `combinations(it, r)` |
| repeats allowed | `product(it, repeat=r)` | `combinations_with_replacement(it, r)` |

Their sizes are Week 55's formulas: n!/(n − r)!, C(n, r), nʳ and C(n + r − 1, r).

`tee(it, n)` splits one iterator into n independent ones, but it must **store** every item that one copy has seen and another hasn't. If one copy runs far ahead, that's the whole stream in memory: when you can, use a list instead, or restructure so one pass does both jobs.
