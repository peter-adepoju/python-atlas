### Bug hunt: B6: Three bugs

`longest_rising(nums)` should return one longest strictly increasing subsequence in O(n log n), by patience with parent links. It has three bugs: it uses `bisect_right` (so equal values extend a run), it never records each element's parent, and it returns the subsequence backwards. Fix them so this prints `[2, 3, 7, 18]`.

```python
import bisect

def longest_rising(nums):
    tops, tops_idx, parent = [], [], [None] * len(nums)
    for i, x in enumerate(nums):
        k = bisect.bisect_right(tops, x)
        if k == len(tops):
            tops.append(x)
            tops_idx.append(i)
        else:
            tops[k], tops_idx[k] = x, i
    out, i = [], tops_idx[-1] if tops_idx else None
    while i is not None:
        out.append(nums[i])
        i = parent[i]
    return out

print(longest_rising([10, 9, 2, 5, 3, 7, 101, 18, 7]))
```
