> **Rules for ties.** Most of the specification is about **ties**: what happens when a job finishes at the same moment another arrives, which of two equally urgent jobs goes first, which free machine gets the job. A simulation without these rules still runs, but two correct-looking programs give different answers, and nobody can say which is right. Writing the rules down, and encoding them in the heap entries themselves (kind before counter, priority before arrival before name), is what makes the simulation **deterministic**: the same input always gives the same log, which is what lets the checks compare you with a completely different simulation. The same thinking applies whenever you use a heap: decide what decides a tie, and put it in the tuple.

## Forge: a cluster simulator

A small computing cluster has `machines` identical machines, numbered from 0. Jobs arrive over time, each with a duration and a priority (smaller is more urgent). Build an **event-driven simulation**: rather than ticking a clock minute by minute, keep a heap of future events and jump from one to the next.

Write `Cluster(machines)` (`ValueError` if fewer than 1) with:

- `submit(time, name, duration, priority)`: a job named `name` will arrive at `time`. Raise `ValueError` for a name already submitted, a negative time or a duration that isn't positive.
- `cancel(time, name)`: at `time`, if the job is **waiting** (arrived and not started), it's removed and never runs; otherwise (not yet arrived, running, finished, or unknown) nothing happens.
- `run()`: simulate everything and return the **log**, a list of `(time, what, name, machine)` tuples, where `what` is `"start"`, `"finish"` or `"cancel"` (with `machine` `None` for a cancel).

The rules, which decide every tie:

1. Events at the same time are handled in this order: all **finishes** (in machine-number order), then all **cancellations**, then all **arrivals** (cancellations and arrivals each in the order `submit` and `cancel` were called). So a job that finishes at 5 frees its machine for a job that arrives at 5, and a cancellation at 5 can't touch a job arriving at 5.
2. After all the events at one time have been handled, **dispatch**: while a machine is free and a job is waiting, start the most urgent waiting job (smallest priority, then earliest arrival, then name) on the **lowest-numbered** free machine. It finishes at the current time plus its duration.

After `run()`:

- `waits()`: a dict from each job that started to how long it waited (start minus arrival);
- `busiest(k)`: the k machines with the most total running time, as `(machine, busy_time)` pairs, busiest first, ties to the lower number;
- `longest_waits(k)`: the k longest waits as `(name, wait)` pairs, longest first, ties alphabetical.

Use heaps for everything that needs a "next": the events, the waiting jobs (with lazy deletion for cancellations) and the free machines. The checks compare you with a simple minute-by-minute simulation on random workloads, and run 20,000 jobs on 8 machines.
