### Bug hunt: An item packed twice

Going through the weights upwards lets the same item be added again and again, so this finds `30` (three flasks) for a limit of 6 with one flask worth 10. Go downwards so each item counts once and it prints `10`.

```python
def pack(items, limit):
    best = [0] * (limit + 1)
    for _, weight, value in items:
        for w in range(weight, limit + 1):
            best[w] = max(best[w], best[w - weight] + value)
    return best[limit]

print(pack([("flask", 2, 10)], 6))
```

```python
# Check: run the exercise cell above first, then this one.
check("def test_output():\n    \"down, not up\"\n    expect_output(\"10\")")
```

## Warm-up

The number of ways to climb `n` steps taking 1, 2 or 3 at a time is the sum of the ways to reach the three steps below. That sentence is the whole solution: a **state** (`n`), a **transition** (the sum), and **base cases** (0 steps: one way; below 0: none).

## The four questions

1. **State**: what does one entry of the table mean? "`best[i]` is the most you can collect from the first `i` houses." Getting this sentence right is most of the work.
2. **Transition**: how does an entry follow from smaller ones? "Either skip house `i` (`best[i-1]`) or take it (`best[i-2] + value`)."
3. **Base cases**: the smallest states, answered directly.
4. **Order**: fill so every entry's inputs are ready (usually from small to large). Or write it top-down with memoisation (`lru_cache`) and let recursion find the order, if the depth is safe.

Then say the complexity: entries times the work per entry. Often you can keep only the last row or two, and interviewers like hearing that.
