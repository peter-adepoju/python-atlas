```python
nums = [4, 2, 6, 1, 5]
answer, stack = [-1] * len(nums), []
for i, x in enumerate(nums):
    while stack and nums[stack[-1]] < x:
        answer[stack.pop()] = x
    stack.append(i)
print(answer, [nums[i] for i in stack])
```

> **Undo that knows its limits.** Week 36's undo assumed nothing else changes the document between a command and its undo. Here the printer works on the queue too, so `undo` can't blindly reverse: it checks that the world still looks the way it left it (the job is still waiting, at the end it joined) and refuses otherwise. That's the honest version of undo in any shared system: an undo is a request, not a guarantee, and it must check before it acts (Week 37's defensive programming). Notice also how little each structure had to do: the ring only ever touches its two ends, the undo stack only its top, the history only overwrites its oldest slot. Every operation is O(1), and the restrictions are what make that possible.

## Forge: a print shop

A small print shop takes jobs from customers. Jobs wait in a queue; an **urgent** job jumps to the front; the printer takes jobs from the front; a customer who changes their mind can **undo** their last submission, if it hasn't been printed yet; and the screen shows the last few jobs printed. Build `PrintShop(capacity=8, keep=5)`.

**Storage.** Waiting jobs are job ids in a **ring deque** on a fixed block of `capacity` slots (Day 5): normal jobs join the back, urgent ones the front. A dictionary maps each id to its `(name, pages)`. Printed ids go into a **ring log** of the last `keep` jobs (it may overwrite the oldest). Each submission is pushed onto an **undo stack** as `(id, end)`, where `end` is `"front"` or `"back"`.

**Operations.**

- `submit(name, pages, urgent=False)`: `pages` must be a whole number of at least 1 (else `ValueError`); a full queue raises `OverflowError("queue is full")`. Returns the new id: `"J1"`, `"J2"`, ... (ids are never reused, even after an undo).
- `print_next()`: removes the front job and returns `(id, name, pages)`; nothing waiting raises `IndexError("nothing to print")`.
- `undo()`: pops the undo stack. If that job is still waiting **at the end it was added to**, remove it and return its id. Otherwise (it's been printed), return `None`. With nothing to undo, return `None`.
- `waiting()`: the waiting ids, front to back. `recent()`: the printed ids, newest first (up to `keep`). `len(shop)`: how many are waiting.

**Interface.** `handle(line, shop)` returns a reply:

| Line | Reply |
|---|---|
| `submit NAME PAGES` | `queued J1` |
| `urgent NAME PAGES` | `queued J2 at the front` |
| `print` | `printed J2 poster (3 pages)` |
| `undo` | `removed J1`, or `nothing to undo` |
| `queue` | `queue: J3 J1` (front first), or `queue: empty` |
| `recent` | `recent: J2 J5` (newest first), or `recent: none` |

`PAGES` is read with `int`; any `ValueError`, `OverflowError` or `IndexError` from the shop becomes `error: ` and its message; anything else gets `unknown command`.
