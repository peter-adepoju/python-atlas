> **Key idea.** `class Name:` makes a class; calling it makes an instance, a new object each time. Attributes are names on an object (`obj.attr`), created by assignment, best all set up in `__init__`. Instances are shared by reference: changes through one name show through all; copy with a new instance or `copy`. By default, `==` means "the same object", and printing shows the class and an address.

## Warm-up

The smallest possible class has an empty body (`pass`). Even so, calling it makes new, separate objects, and each can be given attributes of its own. `q` never got an `x`, so `hasattr(q, "x")` is False.

## The class statement

```
class Name:
    body
```

- The **name** is written in **CapWords** (also called PascalCase): `Account`, `ShoppingBasket`, `HTTPRequest`. Functions and variables stay in `snake_case`, so a capital letter tells a reader "this is a class".
- The **body** is usually method definitions, indented. It can also hold plain assignments (Day 5 is about those), a docstring, or `pass` if there's nothing yet.
- Running a `class` statement **makes a class object** and gives it a name, just as `def` makes a function object. The class is a value: you can print it, store it, pass it around.

The body runs **once**, when the class statement runs. The methods inside are defined then, but don't run until called on an instance.

## Instances and attributes

**Calling** a class makes a new **instance**: a new object whose type is that class. Each call makes another one.

An **attribute** is a name attached to an object, reached with a dot: `rex.name`. You can create one by assigning to it, read it, change it, and (rarely) delete it:

`getattr(obj, "name", default)` reads an attribute whose name is in a string, with a default if it's missing; `hasattr` asks whether it exists. They're handy when the attribute name comes from data.

Adding attributes from outside, as above, works, but it's fragile: an instance only has the attributes someone remembered to add, and a `Dog` without a `name` crashes when it speaks. That's what `__init__` is for: it gives **every** instance the same attributes as it's made. Tomorrow is all about it; here's the shape:

Instances work anywhere other values do: in lists, as dictionary values, as the key function's argument in `max` or `sorted`.

## Instances are shared, not copied

Assigning an instance to another name, putting it in a list, or passing it to a function never copies it. Every one of those is another reference to **the same object**, exactly like lists in Phase 1. So a function that changes an instance's attributes changes it for everyone.

`add_one` changes the object both names refer to. `replace` only makes its own local name `box` refer to a new object; the caller's `b` is untouched. The rule is the same one you learned for lists: **changing an object is visible through every reference; rebinding a name is not.**

To get a separate copy, make a new instance yourself (`Box(list(b.items))`), or use the `copy` module: `copy.copy(b)` makes a **shallow** copy (a new `Box` whose `items` is the **same** list), `copy.deepcopy(b)` copies everything inside too.
