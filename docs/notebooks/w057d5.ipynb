### Bug hunt: B7: Three bugs

`first_at_least(nums, target)` should return the index of the first value at least `target` in a sorted list (`len(nums)` if none), by binary search; `count_between(nums, lo, hi)` counts values in lo to hi inclusive using it. There are three bugs: the search moves `hi = mid - 1`, which can skip the answer; the loop runs `while lo <= hi` with that, which can run past the end; and `count_between` searches for `hi` instead of `hi + 1`, missing values equal to hi. Fix them so this prints `[3, 4, 3]`.

```python
def first_at_least(nums, target):
    lo, hi = 0, len(nums)
    while lo <= hi and lo < len(nums):
        mid = (lo + hi) // 2
        if nums[mid] >= target:
            hi = mid - 1
        else:
            lo = mid + 1
    return lo

def count_between(nums, lo, hi):
    return first_at_least(nums, hi) - first_at_least(nums, lo)

nums = [1, 3, 3, 5, 8, 9]
print([first_at_least(nums, 4), first_at_least(nums, 8), count_between(nums, 3, 5)])
```
