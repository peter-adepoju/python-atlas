## Warm-up

Escaping first means `<` in the text can't become a tag; then the Markdown syntax (backticks, brackets) is turned into tags, which are safe because they were added by the converter, not taken from the text. The order matters: escape, then add markup.

## Static sites

A **static site** is a folder of HTML files, built once and served as they are: no server code, no database, nothing to hack or keep running. Generators (MkDocs and Sphinx for documentation, and many more, all free) turn Markdown into HTML with a template. GitHub Pages serves a folder from a repository for free, which is how this course is published. The build is a program like any other, so it can be tested: that every page renders, every link resolves, and every code example runs (Day 1's checker).

## From Markdown to HTML

A small useful subset: headings (`#`, `##`, `###`), paragraphs separated by blank lines, fenced code blocks (with a language for highlighting), inline code, and links. Give each heading an **anchor** (a slug of its text, made unique with `-2`, `-3` for repeats), so a contents list and other pages can link straight to a section. Escape everything that came from the text.

## Links that work

Broken links are the commonest fault of documentation sites. Before publishing, collect every `href` that points inside the site (not starting with `http`), resolve it against the page it's on, and check the target exists, and, if it has a `#fragment`, that the target page has that anchor. A **sitemap** (`sitemap.xml`, one `<url><loc>...</loc></url>` per page) tells search engines what's there.
