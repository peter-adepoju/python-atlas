````python
class LibraryError(Exception):
    """Something went wrong in the library."""

class NotOnShelf(LibraryError):
    """The book exists but is out on loan."""

class LimitReached(LibraryError):
    """The member already has the most loans allowed."""

def borrow(loans, title, on_shelf, limit=3):
    if len(loans) >= limit:
        raise LimitReached(f"already {len(loans)} loans")
    if title not in on_shelf:
        raise NotOnShelf(f"{title!r} is out")
    on_shelf.remove(title)
    loans.append(title)

shelf = ["Aké", "Small Island", "Purple Hibiscus"]
mine = []
for title in ["Aké", "Aké", "Small Island", "Purple Hibiscus", "Things Fall Apart"]:
    try:
        borrow(mine, title, shelf)
        print("borrowed", title)
    except NotOnShelf as err:
        print("sorry:", err)
    except LibraryError as err:
        print(type(err).__name__, err)
print(mine)
````

## Warm-up

`Ebook` has an empty body, yet its objects have an `__init__`, a `label` method and a `loan_days`: all of them come from `Book`. An ebook also counts as a `Book` for `isinstance`, though its exact type is `Ebook`. That's inheritance, and today is about how it works and when it helps.

## One class built on another

Writing a class name in brackets after a new class makes the new class **inherit** from it:

```python
class Ebook(Book):
    ...
```

Everything `Book` defines (methods, class attributes, special methods like `__init__` and `__repr__`) is available to `Ebook` objects, without copying a line. You'll meet three pairs of words for the two classes; they mean the same thing:

| `Book` | `Ebook` |
|---|---|
| parent class | child class |
| base class | derived class |
| superclass | subclass |

The child then **adds** what's new and **replaces** what's different:

- `Ebook` inherits `__init__`, so `Ebook("Small Island", "Andrea Levy")` works exactly like making a `Book`.
- `label` and `due` are written once, in `Book`. When they run for an ebook, `self` **is** the ebook, so `self.loan_days` finds the ebook's 14. A parent's method automatically uses whatever the child changed.
- `file_name` exists only for ebooks. The parent knows nothing about its children: a plain `Book` has no `file_name`.

## Finding attributes through the family

Week 28, Day 5 gave the rule for reading `obj.name`: look in the instance, then in its class. It left a gap for this week. Here's the full rule:

1. the instance's own attributes;
2. its class;
3. that class's parent, then the parent's parent, and so on up the family;
4. `AttributeError` if it's nowhere.

The first place that has the name wins, so a child's value hides its parent's, and an instance's own value hides both:

## isinstance, issubclass and type

Because every ebook is a book, `isinstance(e, Book)` is True for an ebook. `isinstance` asks "is this object one of these, **or of any class built on it**?" Its partner `issubclass` asks the same question about two classes:

- `isinstance(x, (A, B))` with a tuple is True if `x` is either.
- A class counts as a subclass of itself, so `issubclass(Book, Book)` is True.
- `type(x) == Book` is True **only** for exact `Book` objects. Code that checks the type that way turns away ebooks, which is almost never what you want. Prefer `isinstance`, so that children are welcome wherever their parent is.

When an `if` chain sorts objects with `isinstance`, test the **most specific** class first. An ebook passes `isinstance(e, Item)`, so a chain that checks `Item` first never reaches `Book` or `Ebook`. You'll fix exactly that bug in the practice.

That's why `sum(mark >= 50 for mark in marks)` counts passes: each comparison gives a `bool`, and bools add up like 1s and 0s. ## A family of exceptions

You've been using inheritance since Week 11 without knowing it: exceptions are classes, arranged in families.

`except` uses the same "is a" rule as `isinstance`: `except LookupError` catches both `KeyError` and `IndexError`, because both are lookup errors, and `except Exception` catches nearly everything (which is why it's usually too broad).

For your own programs, define exceptions by inheriting from `Exception`. A class body needs no more than a docstring:

- The message you pass when raising comes back from `str(err)`, as with Python's own exceptions.
- Callers can catch one kind precisely (`except NotOnShelf`) or the whole family (`except LibraryError`), and errors from elsewhere still go past.
- Python tries `except` clauses in order and uses the first that matches, so put the specific ones first. With `except LibraryError` first, the `NotOnShelf` clause would never run.

If your error really is a kind of bad value, inherit from `ValueError` instead (`class BadISBN(ValueError): ...`). Then code that already catches `ValueError` keeps working, and code that wants only your error can catch it by name.
